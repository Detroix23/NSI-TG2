### I- Parcours séquentiel d'une liste

#### 1 - Rappels

Le code ci-dessous va nous permettre de tirer aléatoirement les éléments d'une liste `L`.

````python
from random import randint
L = [randint(1,100) for i in range(20)] # Liste en compréhension. Tire une liste aléatoire.
print("L est une liste aléatoire de 20 éléments:\n",L)
````

Lorsque l'on veut parcourir (et afficher) un à un les éléments d'une liste `L`, on peut :  
* soit **itérer sur les éléments** de la liste ;  


````python
for element in L:
    print(element,end=' ')
````

Dans cette boucle `for`, la variable `element` prend successivement chaque valeur des élements de la liste. 

* soit **itérer sur les indices**. 

````python
for i in range(len(L)):
    element = L[i]
    print(element,end=' ')
````

Dans cette boucle `for`, la variable `i` prend successivement les valeurs `0, ...,len(L)-1` où `len(L)` est la longueur de la liste `L`. Dans cet exemple, `i`prend donc successivement les valeurs 0, 1, 2, ..., 19.

#### 2 - Des exemples



Dans ce paragraphe, les exemples utilisent des listes de personnes.   
Une personne est modélisée par un tuple `(nom, âge)`, par exemple `('Albert',14)`. 

* **Algorithmes de cumul**

*Exemple de calcul d'une moyenne*
    
Voici une fonction qui renvoie la moyenne d’âge, arrondie à l’entier inférieur, d’une liste de personnes.

````python
def moyenne(liste_personnes):
    somme = 0
    for (nom, age) in liste_personnes:
        somme = somme + age
    return somme // len(liste_personnes)

assert moyenne([('Zoe', 17), ('Lea', 19), ('Leo', 12)]) == 16
````

*Exemple de comptage*

Voici une fonction qui compte le nombre de personnes d’une liste qui ont strictement moins de 15 ans.


````python
def nb_enfants(liste_personnes):
    compteur = 0
    for (nom, age) in liste_personnes:
        if age < 15:
            compteur = compteur + 1
    return compteur

assert nb_enfants([('Ano', 14), ('Lea', 19), ('Leo', 12)]) == 2
````

* **Recherche d'un extremum (maximum ou minimum)**

Voici une fonction qui renvoie le nom de la personne la plus âgée d’une liste.


````python
def le_plus_vieux(liste_personnes):
    (nom_vieux, age_max) = liste_personnes[0]
    for (nom, age) in liste_personnes:
        if age > age_max:
            (nom_vieux, age_max) = (nom, age)
    return nom_vieux

assert (le_plus_vieux([('Zoé', 17),('Léa', 19),('Léo', 12)]) == 'Léa')
````

* **Algorithmes de vérification**

On parcourt la liste en faisant une recherche/vérification à chaque étape. On s’arrête dès qu’on trouve ce que l'on cherche ou un contre-exemple dans le cas d’une vérification.


Voici une fonction qui renvoie l’**indice** de la première personne qui a au moins 15 ans dans une liste (renvoie None si aucune personne ne correspond au critère).

````python
def premier_ado(liste_personnes) :
    for i in range(len(liste_personnes)):
        (nom, âge) = liste_personnes[i]
        if âge >= 15:
            return i
    return None

assert premier_ado([('Léo', 12), ('Léa', 19), ('Zoé', 17)]) == 1
assert premier_ado([('Léo', 12), ('Zu', 11), ('Dora', 14)]) == None
````

Ici, nous utilisons `return` en milieu de boucle pour **quitter la fonction**.

* **Exercice 2 : calculer la moyenne des éléments d'une liste**

Compléter le corps de la fonction `moyenne()` ci-dessous. 

````python
def moyenne(ma_liste:list)->float:
    """ Renvoie la moyenne des éléments d'une liste
        Précondition: ma-liste n'est pas vide
    """
    somme = 0
   
    for i in range(len(ma_liste)):
        somme += ma_liste[i]
    moy = somme / len(ma_liste)
   
    return moy 

assert moyenne([-1.5, 0.0, 1.5, 4.0]) == 1.0, 'Le test n\'est pas passé'
L = [randint(1,10) for i in range(5)] # Tire une liste aléatoire
print(L)
moyenne(L)
````

1) Compléter le corps de la fonction `maximum()` ci-dessous.   
**Consigne :** ne pas utiliser la fonction `max()` de Python. 

````python
def maximum(ma_liste:list)->float:
    """ Renvoie le plus grand élément de la liste ma_liste
        Précondition: ma_liste n'est pas vide
    """
    m = ma_liste[0] # Initialise la maximum au premier élément de la liste
    for i in range(len(ma_liste)):
        if m < ma_liste[i]:
            m = ma_liste[i]
    return m

assert maximum([1,2,5,3])==5, 'La fonction maximum ne renvoie pas le bon résultat sur une liste d\'entiers'
assert maximum([1.8,2.4,5.1,3.1])==5.1, 'La fonction maximum ne renvoie pas le bon résultat sur une \
                                        liste de flottants'
L = [randint(1,100) for i in range(1,20)] # Tire une liste aléatoire
print(L)
maximum(L)
````

````python
from random import randint

def maximum_position(ma_liste:list)->tuple:
    """ Renvoie le plus grand élément de la liste ma_liste et son indice
        Précondition: ma_liste n'est pas vide
    """
    m = ma_liste[0] # Initialise la maximum au premier élément de la liste
    for i in range(len(ma_liste)):
        if m < ma_liste[i]:
            m = ma_liste[i]
            ind = i
    return (m,ind)

assert maximum_position([1,2,5,3])==(5, 2), 'La fonction maximum ne renvoie pas le bon résultat sur une liste \
                                                d\'entiers'
assert maximum_position([1.8,2.4,5.1,3.1])==(5.1, 2), 'La fonction maximum ne renvoie pas le bon résultat sur \
                                                une liste de flottants'
L = [randint(1,100) for i in range(1,20)] # Tire une liste aléatoire
print(L)
maximum_position(L)
````

### II - Recherche dichotomique dans une liste triée

La recherche d'un élément dans une liste fait partie des problèmes fréquents.  
La recherche par balayage est la méthode la plus intuitive : on essaie toutes les valeurs jusqu'à trouver la bonne.   
Cependant, lorsque la liste est triée, la recherche dichotomique est beaucoup plus efficace que la recherche séquentielle.

**1 - Introduction : trouver un nombre**

**Exercice :** 

1) Ecrire un programme qui tire un nombre entier aléatoire entre 0 et N, puis qui demande à l’utilisateur d’entrer un nombre jusqu’à ce que la réponse convienne. 


En cas de réponse supérieure au nombre aléatoire tiré, on fera apparaître un message : « ***Plus petit !*** », et inversement, « ***Plus grand !*** » si le nombre est inférieur. Le message « ***Bravo ! Vous avez trouvé le nombre*** » sera affiché à la fin de la partie. 


Rajouter une variable qui compte le nombre de propositions faites avant de découvrir le bon nombre. 


````python
from random import randint

def DevinerNombre(N):
    """
    Entrée : N, un entier
    Sortie : nbCoups, un entier donnant le nombre de coups
    """ 
    nb_mystere = randint(0, N)   # Tirage aléatoire d'un nombre entre 0 et N
    
    proposition = int(input('Votre proposition ? '))    # Première proposition du joueur
    nbCoups = 1                                         # Initialisation du nombre de coups à 1
    
    while proposition != nb_mystere:
        
        if proposition > nb_mystere:
            print('Le nombre proposé est trop grand. ')
            proposition = int(input('Nouvelle proposition : '))

        elif proposition < nb_mystere:
            print('Le nombre proposé est trop petit. ')
            proposition = int(input('Nouvelle proposition : '))

        nbCoups += 1   # Incrémentation du nombre de coups.
            
    print('Vous avez trouvé ! Le nombre mystère est ', nb_mystere)
    print('Nombre de coups : ', nbCoups)  
    return nbCoups

DevinerNombre(100)
````

2) Quelle stratégie permet de trouver le nombre mystère en un minimum de coups ? 

3) Transformer le programme précédent pour que la recherche s'effectue de manière automatique, sans intervention de l'être humain : le programme tire un nombre au hasard entre 0 et N, puis il fait comme s’il avait oublié la valeur à trouver et essaie de deviner ce nombre avec la strategie de la question 2. 


````python
def DevinerNombre_optimise(N):
    """
    Entrée : N, un entier
    Sortie : nbCoups, un entier donnant le nombre de coups
    """
    nb_mystere = randint(0, N)
    elt_min = 0
    elt_max = N
    
    print(nb_mystere)
    
    proposition = N // 2                      # Programme optimisé si proposition = N // 2 (cf. paragraphe suivant)
    nbCoups = 1                               # Initialisation du nombre de coups à 1
    
    while proposition != nb_mystere:
        if proposition > nb_mystere:
            elt_max = proposition - 1    
        
        else:
            elt_min = proposition + 1

        proposition = (elt_min + elt_max) // 2    # Programme optimisé si proposition = proposition // 2 
        nbCoups += 1   # Incrémentation du nombre de coups.

    print('Nombre de coups : ', nbCoups)  
    return nbCoups

DevinerNombre_optimise(100)
````

**2 - Principe de la recherche dichotomique**

La méthode utilisée dans la partie précédente pour deviner un nombre entre 0 et N est appelée **méthode dichotomique** : il s'agit, à chaque étape, de diviser en deux parties l'intervalle de la recherche.

Le terme **dichotomie** vient du grec ancien διχοτομία, dikhotomia (« division en deux parties »).  
Cette méthode  fait partie des méthodes dites **diviser pour mieux régner**.

Extrait de [Wikipedia](https://fr.wikipedia.org/wiki/Recherche_dichotomique) :
![image](diviser_pour_regner.png)
	

La méthode de **recherche dichotomique** nécessite que les valeurs aient été **triées  auparavant**. <br/>
Dans toute la suite, nous rechercherons un élément dans une liste d'entiers **triée** dans l'ordre croissant. 

* ***Premier exemple, en vidéo :*** cliquer [ici](https://www.youtube.com/watch?v=ULr_8ocz0AU) 


Le principe de la recherche dichotomique est donc le suivant :

* Si la liste est vide : répondre négativement, la recherche est finie.


* Sinon, trouver la valeur la plus centrale de la liste et comparer cette valeur à l’élément recherché : 
    * si la valeur est celle cherchée : répondre positivement (ou retourner l'indice de l'élément trouvé), la recherche est finie ;
    * si la valeur est strictement plus petite que l’élément recherché, reprendre la procédure avec la seconde moitié de la liste ;
    * sinon reprendre la procédure avec la première moitié de la liste.

Avec une **recherche séquentielle** ou **recherche par balayage**, on aurait dû parcourir la liste du début à la fin en comparant chaque valeur à l’élément recherché. Dans le pire des cas, comme dans l'exemple précédent, on parcourt la liste en entier : il faudrait faire 9 comparaisons.


Lorsque la liste de départ est triée, la **recherche dichotomique** permet d’améliorer nettement la performance de la recherche.

**3 - L'algorithme**

Voici l'écriture d'un algorithme de recherche dichotomique d'un élément dans une liste d'entiers triée en ordre croissant. Si l'élément est présent, son indice est renvoyé, sinon le booléen `Faux` est renvoyé. <br/>

Nous allons travailler avec deux variables `g` (indice de gauche) et `d` (indice de droite) qui vont délimiter la liste à étudier.  

```python
1. L est une liste d entiers triée, element est un entier 
2. g <- 0
3. d <- dernier indice de la liste
4. tant que g <= d: 
5.     milieu <- (g+d) // 2
6.     si L[milieu] == element:
7.         renvoyer milieu
8.     sinon:  
9.         si L[milieu] > element: 
10.            d <- milieu-1
11.        sinon: 
12.            g <- milieu+1 
13.renvoyer Faux
```

Le programme s'arrête lorsque la valeur cherchée a été trouvée, ou lorsque `d` devient inférieur à `g`.

**4 - Programmation de l'algorithme en Python**

**A vous de jouer...**

````python
def recherche_dicho(L, element) :
    """ 
    Entrées : L, une liste de nombres triée dans l'ordre croissant
              element, l'entier à rechercher dans la liste L
    Sortie : l'indice dans la liste L de l'élément cherché s'il y est, False sinon"""
    
    ind_g = 0
    ind_d = len(L) - 1
    
    while ind_g <= ind_d :
        ind_m = (ind_g + ind_d) // 2
        valeur_centrale = L[ind_m]
        if valeur_centrale == element :
            return ind_m
        elif valeur_centrale > element :
            ind_d = ind_m - 1
        else :
            ind_g = ind_m + 1
            
    return False
    

L = [2, 5, 8, 11, 15, 20, 32]
L2 = [2, 3, 6, 7, 11, 14, 18, 19, 24]

assert recherche_dicho(L, 34) == False
assert recherche_dicho(L2, 14) == 5
````

#### Présentation 
* On commence par chercher, parmi les nombres à trier, un élément plus petit que tous les autres. Cet élément sera le premier de la liste triée : on procède à l'échange entre cet élément et le premier élément de la liste. 
* On cherche ensuite, parmi ceux qui restent, un élément plus petit que tous les autres, qui sera le deuxième du tableau trié : on procède à l'échange entre cet élément et le deuxième élément de la liste. 
* On recommence pour trouver le troisième élément trié et ainsi de suite jusqu'à ce que toute la liste soit entièrement triée.


#### Exemple : 

On cherche à trier la liste suivante : `[29, -6, 12, -11, 10]`

* Le plus petit élement est $-11$, on le place au premier index 0
* Que fait-on du $29$ ? Il prend la place de $-11$, la liste devient donc : `[-11,-6,12,29,10]`
* Le plus petit des élements restants à trier est $-6$, il est déjà bien placé à l'index 1.
* Le plus petit des élements restants à trier est $10$, on le place en troisième position (index 2).
* Que fait-on du $12$ ? Il prend la place de $10$, la liste devient donc : `[-11,-6,10,29,12]`
* Le plus petit des élements restants à trier est $12$, on le place en quatrième position (index 3).
* Que fait-on du $29$ ? Il prend la place de $12$, la liste devient donc : `[-11,-6,10,12,29]`
* Le dernier élément est nécéssairement le plus grand, la liste est triée.

```python
# Recherche du minimum et swap.

4.     si L[j] < L[indexmin]:
```

### IV - Complexité

Au niveau lycée, on s’intéresse presque uniquement à la complexité des algorithmes en temps (il existe aussi une complexité en espace mémoire).   

Il s'agit d'apporter une réponse à la question suivante : 

**Combien d'étapes (au maximum) sont-elles nécessaires pour arriver à la fin de l'algorithme ?** <br/>  


La complexité en temps est notée $\mathcal{O}(fonction)$, où *fonction* est réduite « au plus simple », en prenant uniquement la partie à la croissance la plus rapide, et en supprimant éventuellement les constantes multiplicatives.  

Les complexités les plus fréquentes sont :   
* **Complexité constante $\mathcal{O}(1)$** : le temps d’exécution est constant quelle que soit la taille des données.  
* **Complexité logarithmique $\mathcal{O}(\log_2 n)$**  : schématiquement, si on multiplie la taille des données par 2, alors on augmente le nombre d’opérations de 1.   
* **Complexité linéaire $\mathcal{O}(n)$** : si on augmente la taille des données de $n$, on augmente le temps de traitement de $n$.  
* **Complexité log-linéaire ou quasi-linéaire $\mathcal{O}(n \log_2 n)$** : augmente un peu plus vite que linéaire, mais beaucoup moins que quadratique.  
* **Complexité quadratique $\mathcal{O}(n^2)$**   : si on multiplie la taille des données par 2 (respectivement par 3), on multiplie le temps par 4 (respectivement par 9).   
* **Complexité cubique $\mathcal{O}(n^3)$**   : si on multiplie la taille des données par 2 (respectivement par 3), on multiplie le temps par 8 (respectivement par 27). 
* **Complexité exponentielle $\mathcal{O}(2^n)$ ou $\mathcal{O}(e^n)$** : ce sont des algorithmes **très** lents.

#### 1 - Un exemple d'algorithme de complexité linéaire $\mathcal{O}(n)$

L'algorithme de **recherche par balayage d'un élément dans une liste** est un algorithme dont la complexité est linéaire. 

````python
def recherche_balayage(element,L):
    "si element est dans la liste L, renvoie l'indice de element, sinon renvoie Faux"
    i = 0
    while i < len(L):
        if L[i] == element:
            return i
        i = i + 1
    return False

L=[2, 5, 10, 15, 23, 1]
recherche_balayage(15,L)
````

Le nombre (maximal) d'opérations nécessaires est **proportionnel** à la taille de la liste à étudier. 

#### 2 - Un exemple d'algorithme de complexité logarithmique $\mathcal{O}(\log_2 n)$

L'algorithme de **recherche dichotomique** est un algorithme dont la complexité est logarithmique :  


**Si $n$ est le nombre d'éléments du tableau, la recherche dichotomique se fait, dans le pire des cas, avec un nombre d'étapes correspondant à :** $\boxed{ \left\lfloor \log_{2}(n)\right\rfloor+1 }$.   


* A quoi correspond $\log_2 (n)$ (se lit **logarithme en base 2 de $n$**) ?  
Il s'agit du **nombre de puissances de 2 contenues dans le nombre $n$**. 


* La fonction mathématique $\log_2$ vérifie donc, en particulier, la propriété suivante : pour tout entier $n \geq 0$, $\log_2(2^n)=n$. 


* $\left \lfloor \log_{2}(n) \right \rfloor $ correspond à l'arrondi inférieur du $\log_2 (n)$ . 


* **Exemples :** 
  * Pour $n = 128$ :  
  $log_2(128) = log_2(2^7)= 7$, donc $\left\lfloor \log_{2}(128)\right\rfloor+1 = 7+1 = 8 $.
  
  * Pour $n = 1 \ 000$ :  
  $log_2(1 \ 000) \approx 9,9657$, donc $\left\lfloor \log_{2}(1\ 000)\right\rfloor+1 = 9+1 = 10 $.
  
Ainsi, la recherche dichotomique sur un tableau de 128 valeurs nécessite au maximum 8 étapes.   
Sur un tableau de 1 000 valeurs, elle nécessite au maximum 10 étapes. 


**Exemple détaillé avec un tableau de 16 valeurs :**

Description de la recherche dans le pire des cas (pas de chance, on tombe du côté où il reste le plus de cases à tester !)

* **Etape 1 :** une comparaison sur un tableau trié de 16 cases. On garde 8 cases et on supprime l'élément central et les 7 cases restantes.

* **Etape 2 :** une comparaison sur, au pire, un tableau trié de 8 cases. On garde 4 cases et on supprime l'élément central et les 3 cases restantes.  

* **Etape 3 :** une comparaison sur, au pire, un tableau trié de 4 cases. On garde 2 cases et on supprime l'élément central et la case restante.

* **Etape 4 :** une comparaison sur, au pire, un tableau trié de 2 cases. On garde 1 case et on supprime l'élément central.

* **Etape 5 :** une comparaison sur, au pire, un tableau de 1 case. C'est la dernière case à fouiller.

**Conclusion : autant d'étapes que $log_2(16)+1$ :**  

$log_2(16)+1 = log_2(2^4)+1 = 4 + 1 = 5  $ (dans ce cas, $log_2(16)$ est un entier). 


**Exemple détaillé avec un tableau de 21 valeurs :**

Description de la recherche dans le pire des cas (pas de chance, on tombe du côté où il reste le plus de cases à tester !)

* **Etape 1 :** une comparaison sur un tableau trié de 21 cases. On garde 10 cases et on supprime l'élément central et les 10 cases restantes.

* **Etape 2 :** une comparaison sur, au pire, un tableau trié de 10 cases. On garde 5 cases et on supprime l'élément central et les 4 cases restantes.

* **Etape 3 :** une comparaison sur, au pire, un tableau trié de 5 cases. On garde 2 cases et on supprime l'élément central et les 2 cases restantes.

* **Etape 4 :** une comparaison sur, au pire, un tableau trié de 2 cases. On garde 1 case et on supprime l'élément central.

* **Etape 5 :** une comparaison sur, au pire, un tableau de 1 case. C'est la dernière case à fouiller.

**Conclusion : autant d'étapes que l'arrondi inférieur de $log_2(21)$, auquel on ajoute 1 :**
$\left\lfloor \log_{2}(21)\right\rfloor+1 = \left\lfloor \log_{2}(4,3923)\right\rfloor+1 = 4 +1 = 5$.   

**Remarques :** 
* L'écriture binaire de $16_{10}$ est $10000_2$ : 16 s'écrit avec 5 bits en binaire. 
* L'écriture binaire de $21_{10}$ est $10101_2$ : 21 s'écrit avec 5 bits en binaire. 
On constate que le nombre d'étapes correspond au nombre de bits dans l'écriture binaire. Ceci est un résultat général. 


**A retenir :**   
**De manière générale, si $n$ est le nombre d'éléments du tableau, le nombre maximal d'étapes dans l'algorithme de recherche dichotomique correspond au nombre de bits dans l'écriture binaire de $n$**. 

Autrement dit, $\left\lfloor \log_{2}(n)\right\rfloor+1$ est le **nombre de bits dans l'écriture binaire de $n$**.
