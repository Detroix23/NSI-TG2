## I - Le principe
### Définition

Une **fonction récursive** est une fonction qui s'appelle elle-même.
    
    

### Premier exemple 

Vous avez probablement vu en mathématiques la récursivité lorsque vous avez étudié les **suites définies par récurrence**. On la retrouve aussi comme un puissant moyen de démonstration avec la **démonstration par récurrence** (programme de terminale de la spécialité mathématiques).

Dans ce premier exemple, on s'intéresse à la suite $(u_n)$ définie par son premier terme $u_0 = 2$ et la formule de passage d'un terme au suivant : $ u_n = 3 \ u_{n-1}$. 

On souhaite calculer le terme d'indice $n$ de cette suite. 

```python
def mur(n):
    if n <=1 :
        return 0
    elif (n == 2):
        return 1
    elif (n == 3):
        return 1
    else:
        return mur(n-2) + mur(n-3)    

mur(7)
```

```python
#Pour aller plus loin... 

def liste_murs_longueur_n(n,L):
    """Cette fonction renvoie la liste des murs de longueur n"""
    if n <=1 :
        return L
    elif (n == 2):
        L.append("2")
        return L
    elif (n == 3):
        L.append("3")
        return L
    else:
        liste_murs_longueur_n(n-2,L)
        liste_murs_longueur_n(n-3,L)
            
        for i in range(len(L)):
            longueur = 0
            for j in range(len(L[i])):
                longueur += int(L[i][j])
                            
            if (longueur + 2 == n):
                L.append(L[i]+"2")
            
            elif (longueur + 3 == n):
                L.append(L[i]+"3")
        
        liste_murs = []
        
        for mur in L:
            longueur = 0
            for element in mur:
                longueur += int(element)
            if (longueur == n) and (mur not in liste_murs):
                liste_murs.append(mur)
                          
        return liste_murs
         

liste_murs = liste_murs_longueur_n(10,[])
liste_murs
```

### Version itérative vs version récursive

####  1 - Version itérative

Compléter, de manière logique, la suite de nombres suivante :   
`2    6    18    54    ...     ...    ...     ` 

Dans cette suite de nombres, on dira que :  
* 2 est le terme d'indice 0 et il sera noté $u_0$ ;
* 6 est le terme d'indice 1 et il sera noté $u_1$ ;
* 18 est le terme d'indice 2 et il sera noté $u_2$ ;

#### 2 - Version récursive

La suite de  nombres est donc entièrement définie par la donnée : 
* du premier terme : $u_0 = 2$ ; 
* et d'une relation, appelée **relation de récurrence**, liant un terme et le terme précédent : $u_n = u_{n-1} \times 3$. 

On peut donc également programmer le calcul des termes de cette suite de nombres en s'appuyant sur la remarque suivante :
* Si `n = 0`, alors la fonction renvoie `2` ;
* Sinon, le calcul du terme d'indice `n` s'obtient en multipliant `u(n-1)` par 3.

On obtient donc :

```python
def u(n):
    if n == 0:
        return 2
    else:
        return u(n-1) * 3    # La fonction s'appelle elle-même ici : 
                             # le programme est donc récursif.

assert u(3) == 54
assert u(10) == 118098
```

### Premier exercice

On considère un nouvelle suite de nombres définie par ses deux premiers termes $u_0 = 5$ et $u_1 = 7$ et la formule de récurrence suivante : $ u_n = 2 \times u_{n-2} + 3 \times u_{n-1} $.

1. Compléter la suite de nombres : 

`5    7    31    ...    ...    ...`    

2. Programmer des deux façons précédentes (version itérative et version récursive) le calcul du terme d'indice `n` de cette suite de nombres. Vérifier vos réponses en appelant ces fonctions pour différentes valeurs de `n`.  

#### 1 - Version itérative


```python
def u(n):
    a = 5
    b = 7
    for i in range(n-1):
        c = 2 * a + 3 * b 
        a = b
        b = c
    
    return c

u(2)
```

#### 2 - Version récursive

*On choisit d'appeler la suite $v$ afin de comparer avec les résultats du programme écrit en version itéraive juste avant.* 

```python
def v(n):
    if n == 0:
        return 5
    elif n == 1:
        return 7
    else:
        return 2*v(n-2)+3*v(n-1)
    

assert u(5) == v(5)
assert u(10) == v(10)
```
