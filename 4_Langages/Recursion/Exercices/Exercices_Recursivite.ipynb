<h3 style="color: purple; font-style: 14px italic verdana;"> Longueur d'une liste</h3>
Bien entendu, on n'utilise à aucun moment la fonction len() de python.

```python
def longRec(liste):
    """ Paramètre
    ---------
    liste : (list)
    une liste d'éléments
    
    Résultat
    --------
    (int)
    Cette fonction renvoie le nombre d'éléments de liste
    """
    if not liste:
        return 0
    return 1 + longRec(liste[1:])
```

```python
print(longRec([1, 2, 3]))
print(longRec([2, 3]))
print(longRec([3]))
print(longRec([]))

```

<h3 style="color: purple; font-style: 14px italic verdana;"> Minimum d'une liste</h3>
Bien entendu, on n'utilise à aucun moment la fonction min() de python.

```python
def minimum(a, b):
    return a * int(a < b) + b * int(a > b)

def minRec(liste):
    """
    
    """
    if not liste:
        return None
    if len(liste) == 1:
        return liste[0]
    return minimum(liste[0], minRec(liste[1:]))
    
```

```python
print(minRec([]))
print(minRec([0]))
print(minRec([1, 2]))
print(minRec([3, 2]))
print(minRec([23, 3, 54]))

```

<h3 style="color: purple; font-style: 14px italic verdana;"> Recherche d'une valeur dans une liste</h3>

<p>Bien entendu, on n'utilise à aucun moment la fonction **in** de python.</p>

```python
def rechListeRec(liste, val):
    """
    """
    if not liste:
        return False
    if len(liste) == 1:
        return liste[0] == val
    #print("1", liste[0] == val)
    return (liste[0] == val) | rechListeRec(liste[1:], val)
```

```python
print(rechListeRec([], 4))
print(rechListeRec([1], 4))
print(rechListeRec([4], 4))
print(rechListeRec([1, 4, 3], 4))
```

<h3 style="color: purple; font-style: 14px italic verdana;"> Recherche dichotomique</h3>
On se rappelle que la liste est triée.

Pour la version récursive, une solution est de passer en paramètres la position des pointeurs déterminant le début et la fin de la partie de la liste qu'on examine. Cependant, un utilisateur la pas besoin de savoir qu'on nécessite deux paramètres de plus. C'est pourquoi on va ***cacher*** tout cela dans une fonction de décoration qui ne va faire qu'appeler notre fonction récursive avec les bons paramètres. 

```python
def dichoRec(liste, valeur):
    """
    En regardant ci dessous, compléter les deux derniers paramètres. 
    """
    return dichoRecAux(liste, valeur, 0, len(liste) - 1)


def dichoRecAux(liste, valeur, debut, fin):
    if debut > fin:
        return False
    mid = (debut + fin) // 2
    if liste[mid] > valeur:
        return dichoRecAux(liste, valeur, mid + 1, fin)
    if liste[mid] < valeur:
        return dichoRecAux(liste, valeur, debut, mid - 1)
    return True
```

```python
print(dichoRec([], 5))
print(dichoRec([1], 5))
print(dichoRec([5], 5))
print(dichoRec([1, 5, 10], 5))
print(dichoRec([1, 2, 3, 10], 5))
print(dichoRec([1, 2, 3, 10], 10))
```
