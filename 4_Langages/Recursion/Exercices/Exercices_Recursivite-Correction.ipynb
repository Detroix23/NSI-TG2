<h3 style="color: purple; font-style: 14px italic verdana;"> Longueur d'une liste</h3>
Bien entendu, on n'utilise à aucun moment la fonction len() de python.

```python
def longRec(liste):
    """ Paramètre
    ---------
    liste : (list)
    une liste d'éléments
    
    Résultat
    --------
    (int)
    Cette fonction renvoie le nombre d'éléments de liste
    """
    if liste == []:
        return 0
    else:
        return 1 + longRec(liste[1:])
```

```python
from random import randint

L = [randint(-32,32) for i in range(5)]
print(L)
print(longRec(L))
```

<h3 style="color: purple; font-style: 14px italic verdana;"> Minimum d'une liste</h3>
Bien entendu, on n'utilise à aucun moment la fonction min() de python.

```python
def minRec(liste):
    """
    Paramètre : 
    liste : une liste d'éléments (nombres ou chaînes de caractères)
        
    Renvoie : 
    le plus petit élément de liste. 
    """
    
    if len(liste) == 1:
        return liste[0]
    else:
        elt = minRec(liste[1:])
        if liste[0] < elt:
            return liste[0]
        else: 
            return elt
    
```

```python
from random import randint

L = [randint(-32,32) for i in range(5)]
print(L)
print(minRec(L))
```

<h3 style="color: purple; font-style: 14px italic verdana;"> Recherche d'une valeur dans une liste</h3>

<p>Bien entendu, on n'utilise à aucun moment la fonction **in** de python.</p>

```python
def rechListeRec(liste,val):
    """
    Paramètres : 
    liste : une liste d'éléments (nombres ou chaînes de caractères)
    val : élément recherché dans liste
    
    Renvoie : 
    True si val est dans liste, False sinon.
    """
    if liste == []:
        return False
    elif liste[0] == val:
        return True
    else:
        return rechListeRec(liste[1:],val)
        
L = [8, -30, 2, -10, 21]
assert rechListeRec(L, 50) == False
assert rechListeRec(L, -30) == True
```

<h3 style="color: purple; font-style: 14px italic verdana;"> Recherche dichotomique</h3>
On se rappelle que la liste est triée.

Pour la version récursive, une solution est de passer en paramètres la position des pointeurs déterminant le début et la fin de la partie de la liste qu'on examine. Cependant, un utilisateur la pas besoin de savoir qu'on nécessite deux paramètres de plus. C'est pourquoi on va ***cacher*** tout cela dans une fonction de décoration qui ne va faire qu'appeler notre fonction récursive avec les bons paramètres. 

```python
def dichoRec(liste,valeur):
    """
    En regardant ci dessous, compléter les deux derniers paramètres. 
    """
    return dichoRecAux(liste, valeur, 0, len(liste) - 1)


def dichoRecAux(liste, valeur, debut, fin):
    """
    Paramètres : 
    liste : une liste d'éléments (nombres ou chaînes de caractères)
    valeur : élément recherché dans liste
    debut : indice du premier élément de la portion de liste sur laquelle s'effectue la recherche
    fin : indice du dernier élément de la portion de liste sur laquelle s'effectue la recherche. 
    
    Renvoie : 
    l'indice de l'élément recherché dans liste si l'élément est présent. 
    -1 sinon. 
    """
    
    if debut > fin:
        return -1
    ind_milieu = (debut + fin) // 2
    if valeur == liste[ind_milieu]:
        return ind_milieu
    elif valeur < liste[ind_milieu]:
        return dichoRecAux(liste, valeur, debut, ind_milieu - 1)
    else:
        return dichoRecAux(liste, valeur, ind_milieu + 1, fin)
    
```

```python
from random import randint

L = [randint(-20,20) for i in range(10)]
L.sort()
print(L)

print(dichoRec(L,15))
print(dichoRec(L,L[7]))
```

```python
L = ["lundi","mardi","mercredi","jeudi","vendredi","samedi","dimanche"]
L.sort()
print(L)

print(dichoRec(L,"mercredi"))
print(dichoRec(L,"octobre"))
```

<h3 style="color: purple; font-style: 14px italic verdana;"> Tri par sélection</h3>
Le principe : 

Cas de base : si la liste et vide ou si la liste contient un seul élément, elle est déjà triée.

Sinon, on échange la première valeur de la liste avec la plus petite, puis on rappelle la fonction sur la liste sans la première valeur.

```python
from random import randint

def indice_min(liste):
    """Cette fonction prend en paramètre une liste et renvoie l'indice de la valeur minimale"""
    imin = 0
    for j in range(1,len(liste)):
        if liste[j] < liste[imin]:
            imin = j
    return imin

def echange(liste,i,j):
    """Cette fonction échange les éléments d'indices i et j de la liste liste"""
    tmp = liste[i]
    liste[i] = liste[j]
    liste[j] = tmp
            

def triSelectRec(liste):
    """Version récursive du tri par sélection"""
    if len(liste) == 1 or len(liste) == 0:
        return liste
    else:
        imin = indice_min(liste)
        echange(liste,0,imin)
        
        return [liste[0]] + triSelectRec(liste[1:])
    
L = [randint(-32,32) for i in range(5)]
print(L)
print(triSelectRec(L))  
```

<h3 style="color: purple; font-style: 14px italic verdana;"> Tri par insertion</h3>
Le principe : 

Cas de base : si la liste est vide ou si la liste contient un seul élément, elle est déjà triée.

Sinon, on considère que le début de la liste (jusqu'à l'avant dernière donnée) est déjà triée. Et on va insérer la dernière valeur dans cette liste triée à la bonne place.


```python
def insertion(liste, valeur):
    """
    On considère que la liste est triée par ordre croissant et on veut insérer valeur au bon endroit 
    dans la liste. Pour cela, on insère valeur à la fin de la liste et on permute jusqu'à ce que valeur
    soit arrivée à la bonne place.
    
    """
    pos = len(liste)
    liste.append(valeur)
    
    while pos > 0 and liste[pos-1] > valeur:
        liste[pos] = liste[pos-1]
        pos = pos - 1
        
    liste[pos] = valeur
    return liste
        
    
def triInsertRec(liste):
    if (len(liste) == 1) or (len(liste) == 0):
        return liste
    else: 
        return insertion(triInsertRec(liste[:-1]), liste[-1])
        
L = [randint(-32,32) for i in range(5)]
print(L)
print(triInsertRec(L))

```
