## I - Le principe
### Définition

Une **fonction récursive** est une fonction qui s'appelle elle-même.
    
    

### Premier exemple 

On dispose de briques de longueur 2 et 3.
<img src='images/brique_2.jpg' > <img src='images/brique_3.jpg' >
On souhaite déterminer le nombre de murs distincts, de longueur $n$, que l'on peut construire avec ces briques.

Ecrire une fonction `mur()` qui :
* prend en paramètre un entier `n`, représentant la longueur des murs ;
* renvoie le nombre de murs de longueur `n`. 


```python
def mur(n):
    if n <=1 :
        return 0
    elif (n == 2):
        return 1
    elif (n == 3):
        return 1
    else:
        return mur(n-2) + mur(n-3)    # La fonction s'appelle elle-même ici : 
                                      # le programme est donc récursif.

mur(8)
```

### Version itérative vs version récursive

####  1 - Version itérative

Compléter, de manière logique, la suite de nombres suivante :   
`2    6    18    54    ...     ...    ...     ` 

Dans cette suite de nombres, on dira que :  
* 2 est le terme d'indice 0 et il sera noté $u_0$ ;
* 6 est le terme d'indice 1 et il sera noté $u_1$ ;
* 18 est le terme d'indice 2 et il sera noté $u_2$ ;
* etc. 

Ecrire une fonction `u` qui prend en paramètre un nombre entier `n` et qui renvoie la  valeur du terme d'indice `n` de cette suite de nombres.

```python
def u(n):
    u = 2
    q = 3
    for _ in range(n):
        u *= q

    return u

assert u(3) == 54
assert u(10) == 118098
```

#### 2 - Version récursive

La suite de  nombres est donc entièrement définie par la donnée : 
* du premier terme : $u_0 = 2$ ; 
* et d'une relation, appelée **relation de récurrence**, liant un terme et le terme précédent : $u_n = u_{n-1} \times 3$. 

On peut donc également programmer le calcul des termes de cette suite de nombres en s'appuyant sur la remarque suivante :
* Si `n = 0`, alors la fonction renvoie `2` ;
* Sinon, le calcul du terme d'indice `n` s'obtient en multipliant `u(n-1)` par 3.

On obtient donc :

```python
def u(n):
    if n == 0:
        return 2
    else:
        return u(n-1) * 3    # La fonction s'appelle elle-même ici : 
                             # le programme est donc récursif.

assert u(3) == 54
assert u(10) == 118098
```

directement en renvoyant la valeur 2.

### Premier exercice

On considère un nouvelle suite de nombres définie par ses deux premiers termes $u_0 = 5$ et $u_1 = 7$ et la formule de récurrence suivante : $ u_n = 2 \times u_{n-2} + 3 \times u_{n-1} $.

1. Compléter la suite de nombres : 

`5    7    31    ...    ...    ...`    

2. Programmer des deux façons précédentes (version itérative et version récursive) le calcul du terme d'indice `n` de cette suite de nombres. Vérifier vos réponses en appelant ces fonctions pour différentes valeurs de `n`.  

#### 1 - Version itérative


```python
def u(n):
    u0 = 5
    u1 = 7
    
    if n == 0:
        return u0
    elif n == 1:
        return u1
    else:
        for _ in range(n - 1):
            tmp = u1
            u1 = 2 * u0 + 3 * u1
            u0 = tmp
    
    return u1

print(u(0))
print(u(1))
print(u(2))
print(u(3))
print(u(4))
```

#### 2 - Version récursive

*On choisit d'appeler la suite $v$ afin de comparer avec les résultats du programme écrit en version itérative juste avant.* 

```python
def v(n):
    if n == 0:
        return 5
    elif n == 1:
        return 7
    else:
        return 2 * v(n - 2) + 3 * v(n - 1)

print(v(0))
print(v(1))
print(v(2))
print(v(3))


print(u(5))
assert u(5) == v(5)
assert u(10) == v(10)
```

## II - Les exemples classiques
### 1 - La suite de Fibonacci

La suite de Fibonacci est la suite de nombres définie par : 

<center>
$ \left\{ 
    \begin{array}{lll}
        u_0 = 1\\
        u_1 = 1 \\
        u_n = u_{n-1}+u_{n-2}, pour \  tout \ n \geqslant 2
    \end{array}
  \right. $
    
Écrire une fonction `fib(n)`, définie de façon récursive, qui retourne le terme d'indice $n$ de cette suite de nombres. 

```python
def fib(n):
    if n <= 1:
        return 1
    return fib(n - 2) + fib(n - 1)
    
    
assert fib(4) == 5
assert fib(5) == 8
```

A présent, dessinons l'**arbre d'appels** de cette fonction : 
<center>

</center>

Les instructions `fib(1)` et `fib(0)` sont appelés les **cas de base** de ce programme récursif : il s'agit des cas pour lesquels on peut obtenir le résultat sans avoir recours à l'appel de la fonction. 

Les cas de base sont indispensables à toute fonction récursive : ils permettent au programme de se terminer. En effet, comme une fonction récursive s'appelle elle-même, alors, sans ces cas de base, la fonction s'appellerait à l'infini. C'est pour cette raison que les cas de base sont également appelés **conditions d'arrêt**.

Pour concevoir un programme récursif, il faudra toujours se poser la question : **"quel(s) est(sont) le(s) cas de base ?"** 


### 2 - La somme des $n$ premiers entiers

Pour définir la somme des $n$ premiers entiers, on a l’habitude d’écrire la formule suivante :

<center> $0 \ + 1 \ + 2 \ +	... + \ n \ \ \ \ \\ \ $       (1)
</center>

Une solution pour calculer cette somme consiste à utiliser une boucle **for** pour parcourir tous les entiers *i* entre *0* et *n*, en s’aidant d’une variable intermédiaire *S* pour accumuler la somme des entiers de *0* à *i*. 
On obtient par exemple le programme itératif suivant écrit en Python :


```python
def somme(n):
    S = 0
    for i in range(n+1):
        S = S + i
    return S

somme(100)
    
```

S’il n’est pas difficile de se convaincre que la fonction `somme(n)` ci-dessus calcule bien la somme des `n` premiers entiers, on peut néanmoins remarquer que ce code Python n’est pas directement lié à la formule (1).

En effet, il n’y a rien dans cette formule qui puisse laisser deviner qu’une variable intermédiaire `S` est nécessaire pour calculer cette somme... 

Il existe une autre manière d’aborder ce problème : il s’agit de définir la **fonction récursive** `somme(n)` qui, pour tout entier naturel `n`, donne la somme des `n` premiers entiers de la manière suivante :
<center>
$ somme(n) = \left\{ 
    \begin{array}{ll}
        0 & si \  n = 0 \\
        n + somme(n-1) & si \ n > 0
     \end{array}
  \right. $
    
**A vous :**

1) Ecrire la fonction récursive `somme(n)` qui calcule la somme des `n` premiers entiers. 

2) Dessiner l'arbre d'appels de cette fonction pour l'appel `somme(5)`.  

3) Parcourir cet arbre d'appels "à rebours" pour illustrer le fonctionnement de la fonction `somme`, comme cela a déjà été fait plus haut. 

```python
def somme(n):
    if n == 1:
        return 1
    return n + somme(n - 1)

assert somme(4) == 10
assert somme(100) == 5050
```

### 3 - L'opération de puissance $n$-ième d'un nombre $x$

Il s'agit de la multiplication répétée $n$ fois de $x$ avec lui-même, que l'on écrit habituellement ainsi : 

<center>
$x^n = \underbrace{x \times \ldots \times x}_{n \ fois}$
    
avec, par convention, $x^0 = 1$.

</center> 
Pour écrire une version récursive du calcul de $x^n$, on va définir une fonction `puissance(x,n)`  ainsi :

* On commence par déterminer le cas de base à cette opération : il s'agit du cas pour $n = 0$. 
* Pour définir ensuite la valeur de `puissance(x,n)` pour un entier $n$ strictement positif, on utilise le résultat suivant : 

<center>
$ x^n = x \times \underbrace{x \times \ldots \times x}_{n-1 \ fois} = x \times {x^{n-1}}$
</center>

Ainsi, $x^n$ s'obtient en multipliant $x$ par $x^{n-1}$.
Une définition récursive de l'opération de puissance $n$-ième d'un nombre $x$ est alors la suivante : 

<center>
$ x^n = \left\{ 
    \begin{array}{ll}
        1 & si \  n = 0 \\
        x \times x^{n-1} & si \ n > 0
     \end{array}
  \right. $
    
**A vous :**

1) Ecrire la fonction récursive `puissance(x,n)` qui calcule le nombre $x^n$ pour tout entier naturel $n$. 

2) Dessiner l'arbre d'appels de cette fonction lorsque `x = 3` et `n = 5`. 

3) Parcourir cet arbre d'appels "à rebours" pour illustrer le fonctionnement de la fonction `puissance`comme cela a été fait dans la version récursive du premier exemple du paragraphe I. 


```python
def puissance(x, n):
    if n == 0:
        return 1
    elif n == 1:
        return x
    return x * puissance(x, n - 1)

assert puissance(2, 7) == 128
assert puissance(9, 5) == 59049
```

**A noter :**

La définition d'une fonction récursive n'est pas toujours unique. 

Par exemple, la définition récursive de l'opération de puissance 𝑛-ième d'un nombre 𝑥 peut également être :

<center>
$ x^n = \left\{ 
    \begin{array}{lll}
        1 & si \  n = 0 \\
        x & si \  n = 1 \\
        x \times x^{n-1} & si \ n > 1
     \end{array}
  \right. $

L'introduction d'un deuxième cas de base dans cette nouvelle définition a l'avantage d'éviter de faire la multiplication inutile de $x$ par $1$ qui, dans la première définition est faite à chaque appel de la fonction `puissance(x,n)` à partir du moment où $n \geqslant 0$. On pourrait continuer à ajouter des cas de base pour $n = 2$, $n = 3$, etc., mais cela n'apporterait rien à la définition. En particulier, cela ne réduirait pas le nombre de multiplication. 

Voici une troisième définition récursive de cette même fonction qui distingue le calcul de $x^n$ selon la parité de $n$ : 

<center>
$ x^n = \left\{ 
    \begin{array}{lll}
        1 & si \  n = 0 \\
        {\left( x^{n/2} \right)}^2 & si \  n \ est \ pair \\
        {x \times \left( x^{(n-1)/2} \right)}^2 & si \  n \ est \ impair \\
     \end{array}
  \right. $

L'algorithme qui découle de cette définition porte également le nom d' **exponentiation rapide**. Comme son nom l'indique, il s'agit d'un algorithme particulièrement efficace pour calculer rapidement de grandes puissances entières.

    
**A vous :**

1) Ecrire la fonction récursive `puissancev2(x,n)` qui calcule le nombre $x^n$ pour tout entier naturel $n$ selon la deuxième méthode. 

2) On propose ci-dessous la fonction récursive `puissancev3(x,n)` qui calcule le nombre $x^n$ pour tout entier naturel $n$ en utilisant l'exponentiation rapide. 

   * a) Dessiner l'arbre d'appels de cette fonction pour l'appel `puissancev3(3,5)`. 
   
   * b) Parcourir l'arbre d'appels "à rebours" pour illustrer le fonctionnement de la fonction `puissancev3(x,n)`  
   lorsque `n = 5` et `x = 3`. 
   

```python
def puissance(x, n):
    if n == 0:
        return 1
    elif n == 1:
        return x
    return x * puissance(x, n - 1)

assert puissance(2, 7) == 128
assert puissance(9, 5) == 59049

#assert puissancev2(3,5) == 243

def carre(x):
    return x * x

def puissancev3(x,n):
    if n == 0:
        return 1
    else:
        if n % 2 == 0:
            return carre(puissancev3(x,n // 2))
        else:
            return x * carre(puissancev3(x,(n-1) // 2))

def exp_rapide(x, n):
    """
    Function puissance récurs
    """
    if n == 0:
        return 1
    elif n == 1:
        return x
    # Pair.
    elif n % 2 == 0:
        x_pair = exp_rapide(x, n / 2)
        return x_pair * x_pair
    # Impair.
    x_impair = exp_rapide(x, (n - 1) / 2)
    return x * x_impair * x_impair

print(exp_rapide(2, 4))
print(exp_rapide(2, 3))
print(exp_rapide(3, 4))
print(exp_rapide(10, 10))

# assert puissancev3(3,5) == 243
# assert puissancev3(7,4) == 2401
```

### 4 - La fonction factorielle d'un entier naturel $n$

La factorielle d'un entier naturel $n$, noté $n !$ (se lit **factorielle $n$**), est le produit des nombres entiers strictement positifs inférieurs ou égaux à n : 

<center>
    $n! = n \times (n-1) \times \ldots \times 1$

avec, par convention, $0!=1$.

</center> 

On a, par exemple : 
* $ 5!= 5 \times 4 \times 3 \times 2 \times 1 = 120$. 


La fonction factorielle est définie sur l'ensemble des entiers naturels et admet une définition récursive : 

<center>
$ n! = \left\{ 
    \begin{array}{ll}
        1 & si \  n = 0 \\
        n \times (n-1)! & si \ n \geqslant 1
     \end{array}
  \right. $

**A vous :**

1) Ecrire la fonction récursive `factorielle(n)` qui calcule le nombre $n!$ pour tout entier naturel $n$. 

2) Dessiner l'arbre d'appels de cette fonction pour l'appel `factorielle(7)`. 

3) Parcourir cet arbre d'appels "à rebours" pour illustrer le fonctionnement de la fonction `factorielle` lorsque `n = 7`. 


```python
def factorielle(n):
    if n == 1:
        return 1
    return n * factorielle(n - 1)

assert factorielle(5) == 120
assert factorielle(7) == 5040
```

### 6 - Quelques exemples de programmes récursifs sur les chaînes de caractères

#### a) Les palindromes

On appelle palindrome un mot qui se lit dans les deux sens comme "sos" ou "radar". 

Ecrire une fonction récursive `palindrome` qui teste si un mot est un palidrome. 
* Entrée : Un mot (type `str`)

* Sortie : Un booléen égal à `True` si le mot est un palindrome, `False` sinon.  

**Principe :**

*Cas de base* : 

* Si le mot est la chaîne vide, c'est un palindrome.

* Si le mot ne contient qu'une seule lettre, c'est un palindrome.

*Dans les autres cas* :

* Le mot est un palindrome si et seulement si la première et la dernière lettre sont égales et le mot tronqué de ses première et dernière lettres est un palindrome.


```python
def palindrome(mot):
    if len(mot) <= 1:
        return True
    symmetrie = mot[0] == mot[-1]
    return symmetrie & palindrome(mot[1:-1])
    
print("a", palindrome('antilope'))
print("b", palindrome('radar'))

assert palindrome('antilope') == False
assert palindrome('radar') == True
assert palindrome('tat') == True
assert palindrome('kayak') == True
```

#### b) Les anagrammes

Une anagramme d'un mot est un mot s'écrivant avec les mêmes lettres que le mot initial. 

Par exemple : 
* ironique et onirique.

* baignade et badinage.

* estival et vitales.

Ici, nous ne demandons pas que les mots envisagés soient des mots du dictionnaire. Ainsi 'abc' et 'bca' sont deux anagrammes.

Écrire une fonction python récursive `anagramme` qui renvoie la liste des anagrammes d'un mot. 
* Entrée : Un mot (type `str`)

* Sortie : La liste des anagrammes du mot. 

**Principe :**

*Cas de base* : 

* Si le mot est une chaîne vide, la liste des anagrammes est alors constituée de l'unique chaîne vide.

* Si le mot n'a qu'une seule lettre, la liste de ses anagrammes ne contient qu'un seul élément : le mot lui-même. 

*Dans les autres cas* :

***Un exemple :*** 

On peut définir la liste des anagrammes d'un mot à partir de la liste des anagrammes du mot obtenu en enlevant la première lettre. 

Prenons par exemple le cas d'un mot de trois lettres : `'abc'`. 

* La première lettre est `'a'` ;


* Le mot obtenu en enlevant la première lettre est `'bc'` et la liste des anagrammes de `'bc'` est `['bc','cb']`. 


* Pour constituer la liste des anagrammes de `'abc'`, il suffit de placer la lettre 'a' dans toutes les positions possibles des anagrammes de 'bc'. On obtient la liste `['abc', 'bac', 'bca', 'acb', 'cab', 'cba']`. 


* De la même façon, on obtiendra la liste de toutes les anagrammes du mot `'dabc'` en glissant la lettre `'d'` dans toutes les positions possibles de toutes les anagrammes de `'abc'.`

```python
def listeAnagramme(mot) :
    if mot == '' : 
        return []
    elif len(mot) == 1 :
        return [mot]
    else:
        liste = []
        for anagr in listeAnagramme(mot[1:]):            # pour chaque anagramme du mot privé de sa première lettre,
            for k in range(len(mot)) :                   # on obtient un nouvel anagramme du mot de départ en insérant
                                                         # sa première lettre à toutes les positions possibles.   
                liste.append(anagr[:k] + mot[0] + anagr[k:])
        return liste

print(listeAnagramme('say gex'))
print(listeAnagramme('chat'))
```

#### a) Bubble

Un « bubble » est une figure formée d’un cercle et de deux copies de ce cercle ayant subies une réduction d’un facteur 2, ces deux petits cercles étant tangents extérieurement au cercle initial tels que les lignes des centres soient parallèles aux axes du repère. Ces deux petits cercles deviennent à leur tour « cercle initial » pour poursuivre le dessin.

La figure ci-dessous illustre de gauche à droite des bubbles de profondeur 1 à 4.

<img src='images/bubble.jpg' width = 800>

Un programme permettant d'obtenir la figure à l'étape $n$ où $n$ est un entier naturel strictement positif est donné ci-dessous. La fonction récursive `bubble(n,x,y,r)` où la tortue dessine un bubble de profondeur `n` à partir d'un cercle de centre `(x,y)` et de rayon `r`. A chaque étape, le rayon est divisé par 2.


```python
from turtle import * 

try:
    reset()
    
except Terminator:
    pass

def dessine_cercle(x,y,r):
    up()
    goto(x,y-r)
    down()
    circle(r)
    
    return
    
def bubble(n,x,y,r):
    if n == 0:
        return None
    else:
        dessine_cercle(x,y,r)
        bubble(n-1,x+3/2*r,y,r/2)
        bubble(n-1,x,y-3/2*r,r/2)
    
    return

try:
    bubble(5,-200,100,100)
    exitonclick()  #en cliquant dans la fenêtre graphique, on provoque sa fermeture. 
    
except Terminator:
    print("END.")
```

#### b) Le Flocon de Von Koch

```python
from turtle import *

try:
    reset()
    speed(0)
    
except Terminator:
    pass

def Koch(n,l):
    if n==0:
        forward(l)
    else:
        Koch(n-1,l/3)
        left(60)
        Koch(n-1,l/3)
        right(120)
        Koch(n-1,l/3)
        left(60)
        Koch(n-1,l/3)
    return None
       

def flocon(n,l):
    for k in range(3):
        Koch(n,l)
        right(120)
    return None


try:
    flocon(3,10000)

    exitonclick()  #en cliquant dans la fenêtre graphique, on provoque sa fermeture. 
    
except Terminator:
    print("END.")
```

#### c) Le triangle de Sierpinski

Voici un dernier exemple de figure récursive : le triangle de Sierpsinki. 


```python
import turtle as t
from math import sqrt

try:
    t.reset()
    t.speed(0)
except Terminator:
    pass

t.up()
t.goto(-100,-100)
t.down()

def motif(a):
    t.down()
    t.fillcolor("black")
    t.begin_fill()
    t.forward(a)
    t.left(90)
    t.forward(a)
    t.left(135)
    t.forward(a*sqrt(2))
    t.left(135)
    t.end_fill()
    t.up()


def Sierpinski(n,a):
    if n == 0:
        motif(a)
    else:
        Sierpinski(n-1,a/2)
        t.forward(a/2)
        Sierpinski(n-1,a/2)
        t.left(90)
        t.forward(a/2)
        t.right(90)
        Sierpinski(n-1,a/2)
        t.right(135)
        t.forward(a*sqrt(2)/2)
        t.left(135)        

try:
    Sierpinski(3,200)
except Terminator:
    print("End")
```

L'utilité de la **récursivité** est bien visible dans le cas des figures récursives : on obtient des figures assez complexes en seulement quelques lignes de code. Imaginez le nombre de lignes de codes qu'il serait nécessaire d'écrire pour obtenir ces figures en utilisant une programmation itérative...

**Explication :** 

Une partie de l’espace mémoire d’un programme est organisée sous forme d’une pile où sont stockés les **contextes d’exécution** de chaque appel de fonction. Lorsqu'un appel récursif se termine, cela permet à la fonction appelante de retrouver son contexte propre avec ses variables dans l'état où elles étaient avant appel.

Chaque appel récursif consomme donc de la mémoire dans la pile d'exécution. 

Prenons l'exemple de la fonction `somme`.

L’organisation de la mémoire au début de l’appel à `somme(5)` peut être représentée par la pile ci-dessous contenant un contexte d’exécution avec, entre autres, un emplacement pour l’argument `n` initialisé à `5`, ainsi que d’autres valeurs (comme l’emplacement pour la valeur renvoyée par la fonction) qui sont simplement représentées par des $\ldots$ dans le schéma.

<img src='images/img6.jpg' >

Le calcul récursif de `somme(5)` va engendrer une suite d’appels « en cascade » à la fonction `somme`, que l’on peut représenter par l’arbre d’appels suivant :

<img src='images/img9.jpg' >



En ce qui concerne l’organisation de la pile mémoire, un nouvel environnement d’exécution va être alloué dans la pile pour chacun de ces appels. 

Ainsi, juste après le dernier appel à `somme(0)` la pile contient donc les contextes d’exécution pour les six appels à la fonction `somme`, comme représenté par le schéma ci-dessous.

<img src='images/img7.jpg' >

Ainsi, lors de l’exécution d’une fonction récursive, chaque appel récursif conduit à un empilement du contexte dans la pile d’exécution. Lorsque la condition d’arrêt de la récursivité se produit, les différents contextes sont progressivement dépilés pour poursuivre l’exécution de la fonction...

Toutefois, puisque les appels récursifs sont effectués sur une zone de mémoire plutôt limitée, la pile d’exécution ne doit pas dépasser une certaine limite. Par défaut, elle est de 1000 dans l’implémentation courante de Python. Sinon, on obtient l'erreur constatée précédemment. 

Il est cependant possible de modifier le plafond du nombre d’appels :


```python
import sys
sys.setrecursionlimit(3500)

def somme(n):
    if n == 0:
        return 0
    return n + somme(n - 1)

somme(3000)
```

Toutefois, comme indiqué dans la [documentation](https://python.readthedocs.io/en/latest/library/sys.html#sys.setrecursionlimit), pour des raisons de portabilité, on modifiera avec prudence le plafond des appels en l'élevant de façon modérée...


Enfin, un programme récursif peut être plus lent que son équivalent itératif sans pour autant dépasser le nombre d'appels récursifs autorisés. 

Pour illustrer cela, on s'intéresse à nouveau à la suite de Fibonacci. 

Chacun des deux programmes suivants permet de calculer le terme de rang $n$ de cette suite. 

**Version itérative :**

```python
def fib_v1(n) :
    u, u1, u2 = 1, 1, 1
    for i in range(2,n+1) :
        u = u1 + u2
        u1, u2 = u, u1
    return u
```

**Version récursive :**

```python
def fib_v2(n) :
    if n == 0 or n == 1 :
        u = 1
    else :
        u = fib_v2(n-1) + fib_v2(n-2)
    return u
```

**Comparaison des deux versions**

```python
from time import time

debut = time()
print('u(35) = ',fib_v1(35))
print("Temps d'exécution avec la version itérative : ", time()-debut)

debut = time()
print('u(35) = ',fib_v2(35))
print("Temps d'exécution avec la version récursive : ", time()-debut)
```
