## 2. Vocabulaire de la programmation objet
### 2.1 Classe, objet, attributs, méthodes 

Jusqu'ici nous avons employé uniquement le mot «objet». Il convient maintenant d'être plus précis.  


On désignera par **classe** la structure de données définissant la structure commune à tous les objets qui seront créés à partir d'elle. Plus concrètement, nous pouvons percevoir **une classe** comme un moule grâce auquel nous allons créer autant d'objets de **même type** et **de même structure** qu'on le désire.  

Par exemple, pour modéliser n'importe quelle personne, nous pourrions écrire une classe _Personne_ dans laquelle chaque élément (chaque **objet** de la classe) va se distinguer par des **caractéristiques** qui lui sont propres : nom, prénom, couleur des yeux, couleur des cheveux, âge, taille... On appellera ces caractéristiques des **attributs**.

Chaque **objet** de la classe dipose des mêmes actions que l'on appelle les **méthodes**. Par exemple, n'importe quel être humain dispose, a priori, des actions suivantes : manger, boire, dormir, marcher, courir...  

Lorsque l'**on crée un objet**, on réalise ce que l'on appelle une **instance** de la classe, c'est-à-dire que du moule, on en extrait un nouvel objet (réel) qui dispose de ses attributs et de ses méthodes. L'objet ainsi créé aura pour type le nom de la classe.

#### 2.2.2 La méthode constructeur 

A la création, on dira plutôt à la **construction**, de chaque objet de type `Voiture`, on va lui spécifier directement ses attributs :

la **méthode constructeur**, toujours appelée ```__init__()```, est une méthode (une «def») qui sera automatiquement appelée à la création de l'objet. Elle va donc le doter de tous les attributs de sa classe.

#### Ecriture des méthodes dédiées

Si on veut pouvoir manipuler nos objets, il faut ajouter à notre classe les méthodes souhaitées. Par exemple, on ajoute les méthodes `simplifier` `ajouter`, `multiplier` et `egal` en définissant quatre méthodes dans notre classe.

````python
class Fraction :
    def __init__(self, num, den) :
        self.denominateur = den
        self.numerateur = num
    
    def simplifier(self):
        """simplifie une fraction"""
        a = self.numerateur
        b = self.denominateur
        while b != 0:
            r = a % b
            a,b = b,r
        pgcd = a
        num = self.numerateur // pgcd
        den = self.denominateur // pgcd
        frac = Fraction(num,den)  #On construit la fraction simplifiée.
        if (num % den == 0):      #le numérateur est un multiple du dénominateur : frac est donc un entier.
            frac = num // den            
        return frac
    
    def ajouter(self,other):
        """ajoute la fraction other à la fraction self"""
        num = self.numerateur * other.denominateur + other.numerateur * self.denominateur
        denom = self.denominateur * other.denominateur
        result = Fraction(num,denom)
        result = Fraction.simplifier(result)
        return result
        
    def multiplier(self,other):
        """multiplie la fraction self par la fraction other"""
        num = self.numerateur * other.numerateur
        denom = self.denominateur * other.denominateur
        result = Fraction(num,denom)
        result = Fraction.simplifier(result)
        return result
    
    def egal(self, other):
        """Renvoie Vrai si les deux rationnels sont égaux, Faux sinon."""        
        return self.numerateur * other.denominateur == other.numerateur * self.denominateur
               # produit en croix
````

On peut alors accéder à ces méthodes en utilisant également la **notation pointée** sur l'objet auquel s'applique la méthode.

````python
r1 = Fraction(1, 4)
r2 = Fraction(1, 2)
r3 = r1.ajouter(r2)  # on ajoute r2 à r1
r3.numerateur, r3.denominateur
````

````python
r4 = Fraction(3, 4)
r3.egal(r4)  # pour vérifier si r3 = r4
````

> **Remarque** : Vous noterez que l'on fournit toujours un paramètre de moins lors de l'appel à une méthode que dans la définition de la méthode. En effet, le paramètre `self` n'est pas utilisé car il désigne la référence à l'objet auquel s'applique la méthode.

## 3. Méthodes spéciales en Python

### 3.1 La méthode ```__str__()```

La méthode ```__str__()``` peut redéfinir la manière dont l'objet doit s'afficher lors d'un appel à ```print()```.

Observons comment s'affiche un objet de type ```Fraction``` lorsque rien n'a été spécifié sur son affichage.

````python
a = Fraction(3,4)
````

````python
print(a)
````

C'est un peu décevant. Rajoutons donc une méthode ```__str__()``` .

````python
class Fraction :
    def __init__(self, num, den) :
        self.denominateur = den
        self.numerateur = num
    
    def __str__(self):
        return str(self.numerateur)+"/"+str(self.denominateur)
    
````

````python
a = Fraction(3,4)
print(a)
````

Lorsqu'on affiche un résultat, Python fait automatiquement appel à la méthode `__str__` : l'affichage est maintenant nettement plus agréable !

### 3.2 Les méthodes ```__add__()``` et ```__mul__()```


Il est nécessaire de spécifier le comportement des opérateurs arithmétiques (`+`, `-`, `*`, etc.) sur les types définis par le programmeur, ici le type `Fraction`. 

Définissons par exemple une méthode nommée `__add__` et une méthode nommée `__mul__` pour la classe `Fraction`.

````python
r1 = Fraction(1, 4)
r2 = Fraction(1, 2)
r1 + r2
````

````python
r1 = Fraction(1, 4)
r2 = Fraction(1, 2)
r1 * r2
````

Définissons à présent une méthode nommée `__add__` et une méthode nommée `__mul__` pour la classe `Fraction`.

````python
class Fraction :
    def __init__(self, num, den) :
        self.denominateur = den
        self.numerateur = num
    
    def __str__(self):
        return str(self.numerateur)+"/"+str(self.denominateur)

    def __repr__(self):
        return str(self.numerateur)+"/"+str(self.denominateur)
    
    def simplifier(self):
        """simplifie une fraction"""
        a = self.numerateur
        b = self.denominateur
        while b != 0:
            r = a % b
            a,b = b,r
        pgcd = a
        num = self.numerateur // pgcd
        den = self.denominateur // pgcd
        frac = Fraction(num,den)  #On construit la fraction simplifiée.
        if (num % den == 0):      #le numérateur est un multiple du dénominateur : frac est donc un entier.
            frac = num // den            
        return frac
    
    def __add__(self,other):
        """ajoute la fraction other à la fraction self"""
        num = self.numerateur * other.denominateur + other.numerateur * self.denominateur
        denom = self.denominateur * other.denominateur
        result = Fraction(num,denom)
        result = Fraction.simplifier(result)
        return result
        
    def __mul__(self,other):
        """multiplie la fraction self par la fraction other"""
        num = self.numerateur * other.numerateur
        denom = self.denominateur * other.denominateur
        result = Fraction(num,denom)
        result = Fraction.simplifier(result)
        return result
````

Nous pouvons maintenant utiliser les opérateurs  `+` et `*` sur les objets de type `Fraction` : 
* En appliquant l'opérateur `+` sur des objets de type `Fraction`, Python fait automatiquement appel à `__add__`.

* En appliquant l'opérateur `*` sur des objets de type `Fraction`, Python fait automatiquement appel à `__mul__`. 


````python
a = Fraction(3,8)
b = Fraction(5,2)
c = a + b
d = a * b
print(c)
print(d)
````

La modification du comportement d'un opérateur afin qu'il fonctionne avec des types définis par le programmeur s'appelle la **surcharge d'opérateur**. Pour chaque opérateur de Python, il existe une méthode spéciale correspondante, comme `__add__` ou `__mul__`. 

Pour plus de détails, voir http://docs.python.org/3/reference/datamodel.html#specialnames.


### 3.3 La méthode ```__eq__()```

On pourrait être tenté d'afficher une instance ou de tester l'égalité entre deux instances d'une même classe. Par exemple, avec notre classe `Fraction` on aimerait écrire :


````python
r1 = Fraction(1, 2)
r2 = Fraction(1, 2)
r1 == r2
````

Nous ne pouvons nous satisfaire des résultats. Le test d'égalité fait appel à la méthode __eq__ que nous avons besoin de redéfinir pour obtenir des résultats cohérents.

````python
class Fraction :
    def __init__(self, num, den) :
        self.denominateur = den
        self.numerateur = num
    
    def __str__(self):
        return str(self.numerateur)+"/"+str(self.denominateur)

    def __repr__(self):
        return str(self.numerateur)+"/"+str(self.denominateur)
    
    def simplifier(self):
        """simplifie une fraction"""
        a = self.numerateur
        b = self.denominateur
        while b != 0:
            r = a % b
            a,b = b,r
        pgcd = a
        num = self.numerateur // pgcd
        den = self.denominateur // pgcd
        frac = Fraction(num,den)  #On construit la fraction simplifiée.
        if (num % den == 0):      #le numérateur est un multiple du dénominateur : frac est donc un entier.
            frac = num // den            
        return frac
    
    def __add__(self,other):
        """ajoute la fraction other à la fraction self"""
        num = self.numerateur * other.denominateur + other.numerateur * self.denominateur
        denom = self.denominateur * other.denominateur
        result = Fraction(num,denom)
        result = Fraction.simplifier(result)
        return result
        
    def __mul__(self,other):
        """multiplie la fraction self par la fraction other"""
        num = self.numerateur * other.numerateur
        denom = self.denominateur * other.denominateur
        result = Fraction(num,denom)
        result = Fraction.simplifier(result)
        return result
    
    def __eq__(self, other):
        """Renvoie Vrai si les deux rationnels sont égaux, Faux sinon."""        
        return self.numerateur * other.denominateur == other.numerateur * self.denominateur
               # produit en croix
````

On peut désormais utiliser l'instruction classique de test d'égalité avec les objets de notre classe.

````python
r1 = Fraction(1, 2)
r2 = Fraction(1, 2)
r1 == r2
````

> **Que se passe-t-il pour la dernière instruction ?** Python reconnaît qu'il doit tester l'égalité entre deux instances de la classe `Rationnel`. Ce test (`==`) invoque la méthode spéciale `__eq__` de la classe `Rationnel`. Plus précisément, `r1 == r2` appelle `r1.__eq__(r2)` et comme nous venons de définir cette méthode, le résultat est cohérent. On peut désormais tester l'égalité de deux rationnels sans utiliser l'instruction un peu plus lourde `r1.egal(r2)`.

# Bilan

La **paradigme objet** est une autre façon de voir la programmation qui consiste à utiliser une structure de données appelée **classe** qui réunit des données et des fonctionnalités. Les données sont appelées **atributs** et les fonctionnalités sont appelées **méthodes**.

Une **classe** permet de définir un modèle d'objet en spécifiant des attributs et des méthodes. On peut ensuite utiliser cette classe pour fabriquer des objets selon ce modèle.

En Python, on utilise le mot clé `class` pour définir une classe qui devient alors un nouveau type abstrait de données. On peut alors créer de nouveaux objets en appelant le constructeur qui porte le nom de la classe. Les objets ainsi créés s'appellent des **instances** de la classe. 

En Python, la **méthode spéciale `__init__`** est appelée à la **construction d'un nouvel objet**. C'est dans cette méthode que l'on définit les attributs de nos objets.

Les attributs et méthodes d'une instance de classe sont accessibles en utilisant la notation pointée : `objet.attribut` et `objet.methode(arguments)`.
