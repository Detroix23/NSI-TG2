## <center> <i> Tri à bulles </i> </center>

Le **tri à bulles** ou **tri par propagation** consiste à comparer répétitivement les éléments consécutifs d'un tableau, et à les permuter lorsqu'ils sont mal triés.  

Il doit son nom au fait qu'il déplace rapidement les plus grands éléments en fin de tableau, comme des bulles d'air qui remonteraient rapidement à la surface d'un liquide.  

Pour comprendre le fonctionnement du **tri à bulles**, visualiser la capsule disponible [ici](https://dane.web.ac-grenoble.fr/outils-numeriques-0/decoupeur-de-video#h00xRb-L8Zs#0#0#0)
    
Compléter à présent le code de la fonction `tri_bulles` ci-dessous.

```python
def tri_bulles(t):
    """ Paramètre
    ---------
    t : (...)
    une tableau d'éléments non triés
    
    Résultat
    --------
    (list)
    Cette fonction renvoie ...
    """
    l = len(t)
    
    while l > 0:
        for k in range(0, l - 1):
            if t[k] > t[k + 1]:
                t[k], t[k + 1] = t[k + 1], t[k]
        l -= 1
            
    return t
    
```

```python
# Test
print(tri_bulles([1, 52, 6, -9, 12]))
assert tri_bulles([1, 52, 6, -9, 12]) == [-9, 1, 6, 12, 52]
```

```python
# Tests 
from random import randint

L = [randint(-20, 20) for i in range(10)]
print(L)
tri_bulles(L)
```
