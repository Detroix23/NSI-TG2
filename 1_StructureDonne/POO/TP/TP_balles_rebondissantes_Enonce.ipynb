### 3.3 Étude de la vitesse de contamination

1) Effacer, dans la cellule de code **précédente** toutes les instructions qui figurent en dessous du commentaire *Boucle principale*.

````python
def nombre_malades(Liste):
    """compte le nombre de malades"""
    s = 0
    for b in Liste :
        if b.malade == True :
            s += 1
    return s

cpt = 0
liste_abscisses = [0]
liste_nb_malades = [0]

start_time = time.time()

while continuer :
    fenetre.fill([0,0,0])
    for b in liste_balles :
        for b2 in liste_balles:
            if b != b2:
                b.collision(b2)      # On teste la collision entre la balle b et la balle b2 
        b.rebondir()
        b.deplacer()
        b.dessiner() 
    

    pygame.display.flip()   #Rafraîchissement de l'écran
    time.sleep(0.02)
    cpt += 1
    
    if (cpt % 10 == 0):
        end_time = time.time()
        liste_abscisses.append(liste_abscisses[-1]+end_time-start_time)
        liste_nb_malades.append(nombre_malades(liste_balles))
        start_time = time.time()   
        
    # routine pour pouvoir fermer «proprement» la fenêtre Pygame
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            pygame.display.quit()
            continuer = False

plt.xlabel('temps en secondes')
plt.ylabel('nombre de balles contaminées')
plt.plot(liste_abscisses,liste_nb_malades, marker = 'o', c = 'blue',  markersize = 5)
````
