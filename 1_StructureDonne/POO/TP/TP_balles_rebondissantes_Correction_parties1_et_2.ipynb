````python
import pygame
import time
from pygame.locals import *
from random import randint
from math import sqrt

pygame.display.init()

Jeu = True

ecran = (640,480)               # Choix de la taille de l'écran
fenetre = pygame.display.set_mode(ecran)
fenetre.fill([0,0,0])

x = randint(20, 620)                         # Abscisse initiale de la balle
y = randint(20, 460)                         # Ordonnée initiale de la balle

dx = randint(2, 6)                          # Déplacement de la balle en abscisse (abscisse du vecteur vitesse)
dy = randint(2, 6)                         # Déplacement de la balle en ordonnée (ordonnée du vecteur vitesse)

x1 = randint(20, 620)                         # Abscisse initiale de la balle
y1 = randint(20, 460)                         # Ordonnée initiale de la balle

dx1 = randint(2, 6)                          # Déplacement de la balle en abscisse (abscisse du vecteur vitesse)
dy1 = randint(2, 6)                         # Déplacement de la balle en ordonnée (ordonnée du vecteur vitesse)

r = 30                          # Rayon de la balle

# Couleur de la première balle 
couleur = (randint(0,255),randint(0,255),randint(0,255))  

# Couleur de la deuxième balle
couleur1 = (randint(0,255),randint(0,255),randint(0,255))   

def collision(x,x1,y,y1,dx,dx1,dy,dy1):
    if sqrt((x1-x)**2+(y1-y)**2) <= 2*r:
        print("Collision")
        
        # Dans la collision, il faut gérer le cas où les vecteurs vitesse ont
        # soit la même abscisse, soit la même ordonnée. Sinon, le simple échange 
        # des composantes des vecteurs vitesse n'aurait aucun impact.  
        if dx == dx1 :    
            c = dy         
            dy = (-1) * dy1
            dy1 = (-1) * c 
                            
        elif dy == dy1 : 
            d = dy    
            dy = (-1)*dy1
            dy1 = (-1)*d
            
        else :    
            c = dx    #Echange de dx et dx1 (abscisses des vecteurs vitesse)
            dx = dx1
            dx1 = c
        
            d = dy    #Echange de dy et dy1 (ordonnées des vecteurs vitesse)
            dy = dy1
            dy1 = d
    
    return (dx, dx1, dy, dy1)


def rebondir(x,y,dx,dy,r):
    """Si la balle touche un mur, alors elle rebondit"""
        
    if x + r > 640 or x - r < 0:
        dx = -1 * dx
    if y + r > 480 or y - r < 0:
        dy = -1 * dy
        
    return dx, dy 
   
    
def deplacer(x,y,dx,dy):
    """la balle se déplace : on modifie ses coordonnées"""
    x += dx
    y += dy
    return x,y


def dessiner(x,y,r,couleur):
    """dessiner une balle dont le centre a pour coordonnées (x ; y), de rayon r.""" 
    pygame.draw.circle(fenetre,couleur,(x,y),r)
       

        
while Jeu :            
    fenetre.fill([0,0,0])
        
    #Gestion de la collision entre les deux balles
    dx, dx1, dy, dy1 = collision(x,x1,y,y1,dx,dx1,dy,dy1)
    
    #Gestion du rebond de la première balle
    dx, dy = rebondir(x,y,dx,dy,r)   
    
    #Gestion du rebond de la deuxième balle
    dx1, dy1 = rebondir(x1,y1,dx1,dy1,r)  
      
    
    #Déplacement des deux balles
    x, y = deplacer(x,y,dx,dy)
    x1, y1 = deplacer(x1,y1,dx1,dy1)
    
    #Dessin des deux balles
    dessiner(x,y,r,couleur)
    dessiner(x1,y1,r,couleur1)
    
    pygame.display.flip()     # Rafraîchissement de l'écran
    time.sleep(0.01)
    
    # routine pour pouvoir fermer «proprement» la fenêtre Pygame      
    for event in pygame.event.get():
        if event.type == pygame.QUIT:  #Si l'utilisateur ferme la fenêtre
            pygame.display.quit()
            Jeu = False
    
````

````python
import pygame
import matplotlib.pyplot as plt
import time
from pygame.locals import *

from random import randint

ecran = (640,480)

pygame.display.init()
fenetre = pygame.display.set_mode(ecran)
fenetre.fill([0,0,0])

class Balle :
    def __init__(self):
        self.x = randint(0,ecran[0])
        self.y = randint(0,ecran[1])       
        self.dx = randint(-5,5)
        self.dy = randint(-5,5)
        self.malade = False
        self.color = (255,255,255)
        self.taille = 10
        
    def deplacer(self):
        # mouvement de la balle self
        self.x += self.dx
        self.y += self.dy
        
            
    def collision(self,other):    # collision de la balle self avec une autre balle
        if ((other.x-self.x)**2 + (other.y-self.y)**2)**0.5 <= self.taille + other.taille :
            # on échange les dx
            other.dx, self.dx = self.dx, other.dx
            # on échange les dy
            other.dy, self.dy = self.dy, other.dy 
            if (self.malade == True) or (other.malade == True):
                other.malade = True      #S'il y a collision avec la balle other, alors les deux balles sont malades         
                self.malade = True
                
    def dessiner(self):
        if self.malade == True:
            self.color = (255,0,0)
        pygame.draw.circle(fenetre,self.color,(self.x,self.y),self.taille)
    
    def contamine(self,other):
        if (((other.x-self.x)**2 + (other.y-self.y)**2)**0.5 < self.taille + other.taille) and (other.malade == True):
            self.color = (255,0,0)      #S'il y a collision avec la balle other et que celle-ci est malade, la balle devient malade
        
    def rebondir(self):
        """Si la balle touche un mur, alors elle rebondit"""
        if self.x + self.taille >= 640 or self.x - self.taille <= 0:
            self.dx = (-1) * self.dx

        if self.y + self.taille >= 480 or self.y - self.taille <= 0:
            self.dy = (-1) * self.dy


nb_balles = 100
liste_balles = []

for k in range(nb_balles):
    b = Balle()
    liste_balles.append(b)
    
liste_balles[0].malade = True

continuer = True

def nombre_malades(Liste):
    """compte le nombre de malades"""
    s = 0
    for b in Liste :
        if b.malade == True :
            s += 1
    return s

cpt = 0
liste_abscisses = [0]
liste_nb_malades = [0]

start_time = time.time()

while continuer :
    fenetre.fill([0,0,0])
    for b in liste_balles :
        for b2 in liste_balles:
            if b != b2:
                b.collision(b2)      # On teste la collision entre la balle b et la balle b2 
                #b.contamine(b2)      # On teste la collision entre la balle b et la balle b2 
        b.rebondir()
        b.deplacer()
        b.dessiner() 
    

    pygame.display.flip()   #Rafraîchissement de l'écran
    time.sleep(0.02)
    cpt += 1
    
    if (cpt % 10 == 0):
        end_time = time.time()
        liste_abscisses.append(liste_abscisses[-1]+end_time-start_time)
        liste_nb_malades.append(nombre_malades(liste_balles))
        start_time = time.time()   
        
    # routine pour pouvoir fermer «proprement» la fenêtre Pygame
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            pygame.display.quit()
            continuer = False

plt.xlabel('temps en secondes')
plt.ylabel('nombre de balles contaminées')
plt.plot(liste_abscisses,liste_nb_malades, marker = 'o', c = 'blue',  markersize = 5)
````
