## I.  Les piles  

**Rappel de cours :**  
En informatique, une pile (en anglais stack) est une structure de données fondée sur le principe «dernier arrivé, premier sorti » (ou **LIFO** pour **Last In, First Out**), ce qui veut dire que les derniers éléments ajoutés à la pile seront les premiers à être récupérés.

Le fonctionnement est donc celui d'une pile d'assiettes :

On ajoute des assiettes sur la pile, et on les récupère dans l'ordre inverse, en commençant par la dernière ajoutée.  


Source : https://commons.wikimedia.org/w/index.php?curid=21139706  


**Les fonctions "primitives" pour les piles**  

Ces fonctions "primitives" sont les suivantes : 
* création d'une pile vide, 
* tester si une pile est vide, 
* empiler : insérer un élément en haut de la pile,   
* dépiler : renvoyer la valeur de l'élément en haut de la pile ET le supprimer de la pile.

Et rien de plus ...  

On donne ci-dessous une possible implémentation de ces fonctions primaires, en utilisant le vocabulaire de la programmation orintée objet que nous avons déjà abordée. Il existe bien d'autres possibilités pour implémenter ces fonctions primaires.   
Dans toute la suite de ce TD, les piles seront implémentées à l'aide des `list` python. Le **sommet** de la pile sera **l'élément écrit le plus à droite**.  

Ainsi, si l'on part d'une pile vide et que l'on empile successivement les entiers 1, puis 2, puis 3, on obtiendra une pile qui s'affichera de la façon suivante : `[1, 2, 3]`, le sommet de cette pile est l'entier 3.

```python
class Pile :
    def __init__(self):
        self.contenu = []
        
    def est_vide(self) :
        return self.contenu == []
    
    def empiler(self,x):
        self.contenu.append(x)
        
    def depiler(self):
        assert not self.est_vide(), "la pile est vide"
        return self.contenu.pop()
    
    def __repr__(self):       
        ch = ""              
        for x in self.contenu:
            ch = "|\t" + str(x) + "\t|" + "\n" + ch
        ch = "\nEtat de la pile:\n" + ch
        return ch
```

Tester ci-dessous ces primitives :

```python
pile_1 = Pile()
pile_1.empiler("a")
print(pile_1.contenu)
pile_1.empiler("b")
print(pile_1.contenu)
```

## Piles et files

On dispose d’une interface d’un type `Pile` :
* `p = Pile()` permet de créer une pile vide ;
* `p.est_vide()` renvoie un booléen disant si la pile est vide ou pas ; 
* `p.empiler(n)` empile l’élément `n` sur la pile `p` ; 
* `p.depiler()` renvoie l’élément du haut de la pile et le supprime.

On dispose aussi d’une interface d’un type `File` :
* `f = File()` permet de créer une file vide ;
* `f.est_vide()` renvoie un booléen disant si la file est vide ou pas ;
* `f.enfiler(n)` enfile l’élément `n` au bas de la file `f` ; 
* `f.defiler()` renvoie l’élément du haut de la file et le supprime.
