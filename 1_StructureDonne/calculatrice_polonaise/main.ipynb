### <center> <em> Un exemple d'utilisation de la notion de pile en informatique </em>

On donne ci-dessous un script Python qui permet de modéliser la structure de pile.

```python
import typing

Value = typing.Any

class Cellule:
    contenu: Value
    suivante: typing.Optional['Cellule']
    
    def __init__(self, contenu: Value, suivante: typing.Optional['Cellule']) -> None:
        self.contenu = contenu
        self.suivante = suivante
        
    def __repr__(self) -> str:
        return f"Cellule(contenu={self.contenu}, suivante={self.suivante})"

class Pile:
    data: typing.Optional['Cellule']
    
    def __init__(self):
        """Initialise la pile comme une pile vide."""
        self.data = None
    
    def __str__(self):
        """Affiche les éléments de la pile"""
        s: str = ""
        c: 'Cellule' = self.data
        while c != None :
            s += "|\t" + str(c.contenu)+ "\t|" + "\n"
            c = c.suivante
        s = "\nEtat de la pile (début):\n" + s
        return s
    
    def __repr__(self) -> str:
        return f"Pile(data={self.data})"
    
    def __len__(self) -> int:
        length: int = 0
        reserve: Pile = Pile()
        # Comptage.
        while not self.est_vide():
            reserve.empile(self.depile())
            length += 1
        
        # Tu casses, tu répares; tu salis, tu nettoies.
        while not reserve.est_vide():
            self.empile(reserve.depile())
            
        return length
    
    @staticmethod
    def from_arguments(*values) -> 'Pile':
        """
        Crée une pile depuis une list d'arguments.
        """
        pile: Pile = Pile()
        for value in values:
            pile.empile(value)
        
        return pile
    
    def to_list(self) -> list[Value]:
        """
        Renvoie une copie sous type liste.
        """
        reserve: Pile = Pile()
        l: list[Value] = []
        while not self.est_vide():
            element: Value = self.depile()
            reserve.empile(element)
            l.append(element)
            
        while not reserve.est_vide():
            self.empile(reserve.depile())
        
        return l
    
    def est_vide(self):
        """Renvoie True si la liste est vide, False sinon."""
        return self.data == None
    
    def empile(self, x):
        """Ajoute l'élément x sur le sommet de la pile, ne renvoie rien."""
        self.data = Cellule(x, self.data)
    
    def depile(self):
        """Retire l’élément au sommet de la pile et le renvoie."""
        v: typing.Any = self.data.contenu           # on récupère la valeur à renvoyer
        self.data = self.data.suivante  # on supprime la 1ère cellule  
        return v
    
```

```python
# Question 1
p = Pile()
for value in [5, 7, 2, 4]:
    p.empile(value)
    
# Question 2
def pile_contient(pile: Cellule, element: Value) -> bool:
    """
    Paramètres
    ----------
    pile : (list)
      La pile dans laquelle on veut tester la présence d'un élément.
    element : (int ou str)
      L'élément dont on veut tester l'existence ou non dans la pile.
    
    Résultat
    --------
    un booléen égal à True si l'élément est dans la pile, False sinon. 
    """
    checked: Pile = Pile()
    found: bool = False
    # Recherche linéaire.
    while not found and not pile.est_vide():
        value: Value = pile.depile()
        found = value == element
        checked.empile(value)
    
    # Reconstruction.
    while not checked.est_vide():
        pile.empile(checked.depile())
    
    return found
    

# Question 3
def calcule_somme(pile: Pile) -> int:
    """ Cette fonction calcule la somme des éléments de la pile.
    Paramètre : pile (list)
    Sortie : un entier correspondant à la somme des éléments de la pile
    """
    total: int = 0
    checked: Pile = Pile()
    # Somme.
    while not pile.est_vide():
        value: Value = pile.depile()
        total += value
        checked.empile(value)
        
    # Reconstruction.
    while not checked.est_vide():
        pile.empile(checked.depile())
        
    return total

# Question 4 
def at_index(pile: Pile, index: int) -> Value:
    """
    Renvoie l'element de la `pile` à l'`index` donné.  
    0 correspond au sommet.
    """
    reserve: Pile = Pile()
    element: Value = pile.depile()
    reserve.empile(element)
    # Consommation.
    while index > 0 and element is not None:
        element = pile.depile()
        reserve.empile(element)
        index -= 1
    
    # Reconstruction.
    while not reserve.est_vide():
        pile.empile(reserve.depile())
    
    return element

def avant_dernier(pile: Pile) -> Value:
    """ Cette fonction retourne l'avant-dernier élément de la pile.
    Paramètre : pile (list)
    Sortie : un entier correspondant à l'avant-dernier élément de la pile.
    """
    return at_index(pile, len(pile) - 2)
        

```

**Activité 2 : (gare de triage)**

*Objectif : résoudre un problème de triage en modélisant une zone de stockage par une pile.*

En suivant les instructions du chef de gare et en utilisant les opérations sur une pile, écrire la fonction `tri_wagons` qui sépare les wagons bleus et rouges d’un train.

```python
def tri_wagons(train: str) -> str:
    """ Trie les wagons rouges/bleus d'un train
    Paramètre
    ---------
    train : (str) 
      une chaîne de caractères représentant un train avec des wagons bleus (nombre) et rouges (lettres).
    
    Résultat
    --------
    une chaîne de caractères représentant le nouveau train dans lequel les wagons sont triés avec les bleus d'abord et les 
    rouges ensuite. 
    """
    trie: Pile = Pile()
    numeric: Pile = Pile()
    # Sort.
    for wagon in train.split(" "):
        if wagon.isnumeric():
            numeric.empile(wagon)
        else:
            trie.empile(wagon)
        
    while not numeric.est_vide():
        trie.empile(numeric.depile())
        
    # print("Num:", numeric)
    # print("Alpha:", alphabetic)
        
    return " ".join(trie.to_list())
    
```

```python
assert tri_wagons("A 4 C 12") == "4 12 C A", tri_wagons("A 4 C 12")
assert tri_wagons("K 8 P 17 L B R 3 10 2 N") == "8 17 3 10 2 N R B L P K", tri_wagons("K 8 P 17 L B R 3 10 2 N")
```

*Objectif : programmer une mini-calculatrice qui calcule les expressions en écriture polonaise.*

1. Écrire une fonction `operation` qui calcule la somme ou le produit de deux nombres `a` et `b`.


2. Ecrire une fonction `calculatrice_polonaise` qui correspond à l'implémentation, en Python, de l'algorithme écrit en langage naturel dans le document pdf. 


3. Modifier le code pour prendre en charge la soustraction et la division. 



```python
#Question 1 

OPERATORS: set[str] = {"+", "-", "*", "/"}

def operation(a: int, b: int, op: str) -> typing.Union[float, int]:
    """ Calcule la valeur de 'a + b 'ou 'a * b' ...
    Paramètres
    ----------
    a, b : (int) 
      les opérandes
    op : (str) 
      un caractère correspondant à l'opérateur à appliquer aux nombres a et b. 
    
    Résultat
    --------
      le résultat de l'opération (int ou éventuellement float)"""
    
    if op == "+":
        return a + b
    if op == "-":
        return a - b
    if op == "*":
        return a * b
    if op == "/":
        return a / b
    
    raise ValueError(f"(X) - operation - Operator not supported ({op})")

#Question 2
def calculatrice_polonaise(expression: str):
    """ Calcule la valeur d'une expression codée en notation polonaise
    Paramètre
    ---------
    expression : (str)
      une chaîne de caractères représentant une expression en notation polonaise.
    
    Résultat
    --------
      le résultat de l'opération (nombre entier ou flottant)."""

    digits: Pile = Pile()
    
    for symbol in expression.split(" "):
        # print(f"! {symbol=}")
        
        if symbol.isdigit():
            digits.empile(symbol)
        
        elif symbol in OPERATORS:
            a: float = float(digits.depile())
            b: float = float(digits.depile())
            digits.empile(operation(b, a, symbol))
            # print(f"! operation({b}, {a}, {symbol})")
        
        else:
            raise ValueError(f"(X) - calculatrice_polonaise - Unknown symbol ({symbol}).")
        
        # print(f"! {digits}")
        
    return digits.depile()
```

```python
# Tests question 1
assert operation(2, 4, "+") == 6
assert operation(2, 4, "*") == 8
assert operation(8, 4, "/") == 2.0

# Tests question 2
# print(calculatrice_polonaise("2 3 4 + +"))
assert calculatrice_polonaise("2 3 4 + +") == 9.0
assert calculatrice_polonaise("2 3 + 5 *") == 25.0
assert calculatrice_polonaise("8 7 3 + *") == 80.0
assert calculatrice_polonaise("8 7 3 * +") == 29.0
# print(calculatrice_polonaise("3 5 * 2 + 7 * 3 8 1 - * +"))
assert calculatrice_polonaise("3 5 * 2 + 7 * 3 8 1 - * +") == 140.0
assert calculatrice_polonaise("1 2 1 2 1 2 1 2 + * + * + * +") == 31.0
assert calculatrice_polonaise("6 1 5 1 4 1 3 1 2 1 - * - * - * - * -") == 3.0
assert calculatrice_polonaise("1 1 2 1 3 1 2 + / + / + /") == 0.5
```

```python
def parentheses_correctes_compte(expression):
    """
    expression : (str) une expression mathématique en notation classique.
    resultat: un booléen égal à True si l'expression est bien parenthésée, False sinon.
    """
    
    index: int = 0
    count: int = 0
    while index < len(expression) and count >= 0:
        if expression[index] == "(":
            count += 1
        elif expression[index] == ")":
            count -= 1
        
        index += 1
    return count == 0

def parentheses_correctes(expression):
    """
    expression : (str) une expression mathématique en notation classique.
    resultat: un booléen égal à True si l'expression est bien parenthésée, False sinon.
    """
    
    parentheses: Pile = Pile()
    correct: bool = True
    index: int = 0
    while correct and index < len(expression):
        if expression[index] == "(":
            parentheses.empile("(")
        elif expression[index] == ")":
            if parentheses.est_vide():
                correct = False
            else:
                parentheses.depile()
        
        index += 1
    
    correct &= parentheses.est_vide()
    
    return correct
            
def crochets_parentheses_correctes(expression):
    """ 
    Paramètre
    ---------
    expression : (str) 
      une expression mathématique en notation classique.
      
    Résultat
    -------- 
      un booléen égal à True si les parenthèses et les crochets de l'expression sont cohérents, False sinon."""
    OPENING: set[str] = {"(", "["}
    CLOSING: set[str] = {")", "]"}
    
    parentheses: Pile = Pile()
    correct: bool = True
    index: int = 0
    while correct and index < len(expression):
        if expression[index] == "(":
            parentheses.empile("(")
        elif expression[index] == ")":
            if parentheses.est_vide():
                correct = False
            else:
                parentheses.depile()
        
        index += 1
    
    correct &= parentheses.est_vide()
    
    return correct
    
```
