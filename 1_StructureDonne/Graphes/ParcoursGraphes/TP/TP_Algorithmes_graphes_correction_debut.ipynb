## <center> Parcours d'un graphe : correction du TP </center>

```python
class Cellule :
    def __init__(self, contenu, suivante):
        self.contenu = contenu
        self.suivante = suivante
```

```python
class Pile:
    def __init__(self):
        self.data = None
    
    def est_vide(self):
        return self.data == None
    
    def empile(self, x):
        self.data = Cellule(x,self.data)
    
    def depile(self):
        v = self.data.contenu #on récupère la valeur à renvoyer
        self.data = self.data.suivante  # on supprime la 1ère cellule  
        return v
   
```

```python
class File:
    def __init__(self):
        self.data = None
    
    def est_vide(self):
        return self.data == None
    
    def enfile(self, x):
        self.data = Cellule(x,self.data)
    
    def defile(self):
        c = self.data
        if c.suivante != None:    #si la liste chaînée ne contient pas qu'un seul élément
            while c.suivante != None:
                prec = c
                c = c.suivante
            prec.suivante = None
        else:                     #Cas où la liste chaînée contient un seul élément
            self.data = None
        return c.contenu
    
```

```python
graphe_1 = {"A" : ["F","B"],
          "B" : ["C","D","G"],
          "C" : ["B","E"],
          "D" : ["B","I"],
          "E" : ["C","I"],
          "F" : ["A","G","H"],
          "G" : ["F","I","H","B"],
          "H" : ["F","I"],
          "I" : ["D","E","G","H"]
         }

GrapheL = {"A" : ["Z"],
          "B" : ["D"],
          "D" : ["B","G"],
          "E" : ["G","Z"],
          "F" : ["K","N"],
          "G" : ["D","E","N"],
          "K" : ["F","N"],
          "N" : ["F","G","K","Z"],
          "Z" : ["A","E","N"]
         }

GrapheG = {"A" : ["B","C","E"],
          "B" : ["D","F","E"],
          "C" : ["D"],
          "D" : ["E"],
          "E" : ["A"],
          "F" :["E"]
         }
```

```python
def voisins(g, s):
    """
    Renvoie la liste des voisins de s si g est non orienté,
    des successeurs si g est orienté.
    """
    L = g[s]
    L.sort()
    return L

voisins(GrapheL, "G")
```

## Parcours en largeur

```python
def parcours_largeur(G, s):
    f = File()
    f.enfile(s)
    decouverts = [s]
    while not f.est_vide():
        s = f.defile()
        print(s,end=' ')
        for v in voisins(G,s):
            if v not in decouverts:
                decouverts.append(v)
                f.enfile(v)
    #return decouverts
```

```python
parcours = parcours_largeur(GrapheG, 'B')
parcours
```

```python
parcours = parcours_largeur(GrapheL, 'D')
parcours
```

## Parcours en profondeur

```python
def parcours_profondeur(G, s):
    p = Pile()
    p.empile(s)
    visites = []
    while not p.est_vide():
        s = p.depile()
        if s not in visites:
            print(s,end= " ")
            visites.append(s)
            for v in voisins(G,s):
                if v not in visites:
                    p.empile(v)
    #return visites
    
```

```python
parcours = parcours_profondeur(GrapheG,"B")
parcours
```

```python
parcours = parcours_profondeur(GrapheL,"D")
parcours
```

```python
def parcours_profondeur_recursif(G, s,visites):
    visites.append(s)
    print(s, end = " ")
    for v in voisins(G,s):
        if v not in visites:
            parcours_profondeur_recursif(G, v, visites)
    return visites   
    
```

```python
parcours = parcours_profondeur_recursif(GrapheG, "B", [])
```

```python
print(parcours)
```

```python
parcours = parcours_profondeur_recursif(GrapheL, "B", [])
```
