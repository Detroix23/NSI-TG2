### I - Définition des classes `Cellule`, `Pile` et `File`

Il n'y a rien d'autre à faire dans cette partie que d'exécuter les cellules de code : ces classes seront utiles dans la partie suivante pour implémenter les parcours en largeur et en profondeur d'un graphe. 


```python
class Cellule :
    def __init__(self, contenu, suivante):
        self.contenu = contenu
        self.suivante = suivante
        
```

```python
class Pile:
    def __init__(self):
        self.data = None
    
    def est_vide(self):
        return self.data == None
    
    def empile(self, x):
        self.data = Cellule(x,self.data)
    
    def depile(self):
        v = self.data.contenu #on récupère la valeur à renvoyer
        self.data = self.data.suivante  # on supprime la 1ère cellule  
        return v
    
    def __str__(self):
        s = ""
        c = self.data
        while c != None :
            s += "|\t" + str(c.contenu)+ "\t|" + "\n"
            c = c.suivante
        s = "\nEtat de la pile:\n" + s
        return s
   
```

```python
class File:
    def __init__(self):
        self.data = None
    
    def est_vide(self):
        return self.data == None
    
    def enfile(self, x):
        self.data = Cellule(x,self.data)
    
    def defile(self):
        c = self.data
        if c.suivante != None:    #si la liste chaînée ne contient pas qu'un seul élément
            while c.suivante != None:
                prec = c
                c = c.suivante
            prec.suivante = None
        else:                     #Cas où la liste chaînée contient un seul élément
            self.data = None
        return c.contenu
    
    def __str__(self):
        s = ""
        c = self.data
        while c != None :
            s =  "|" + str(c.contenu) + s
            c = c.suivante
        s = "\nEtat de la pile:\n" + s + "|"
        return s
```

### II - Parcours en largeur et parcours en profondeur d'un graphe


```python
# Implémentation de graphe 1 et graphe 2 à l'aide de dictionnaires

G1 = {"A" : ["B","F"],
      "B" : ["A", "C", "D", "G"],
      "C" : ["B", "E"],
      "D" : ["B", "I"],
      "E" : ["C", "I"],
      "F" : ["A", "G", "H"],
      "G" : ["B", "F", "I"],
      "H" : ["F", "I"],
      "I" : ["E", "D", "G", "H"]}


```

```python
def voisins(G, s):
    """ Paramètres
    ---------
    G : (dict)
    un dictionnaire représentant un graphe
    
    s : (string)
    sommet
    
    Résultat
    --------
    (list)
    Cette fonction renvoie la liste des voisins de s si G est non orienté, la liste des successeurs si G est orienté, 
    les sommets étant rangés dans l'ordre alphabétique.
    """
    L = G[s]
    L.sort()
    return L
```

```python
def parcours_largeur(G, s):
    """ Paramètres
    ---------
    G : (dict)
    un dictionnaire représentant un graphe
    
    s : (string)
    sommet de départ 
    
    Résultat
    --------
    None
    Cette fonction ne renvoie rien mais affiche les sommets du graphe au fur et à mesure qu'ils sont atteints 
    par le parcours en largeur.
    """
    f = File()
    f.enfile(s)
    decouverts = [s]
    while f.est_vide() != True :
        s = f.defile()
        print(s)
        for v in voisins(G, s) :
            if v not in decouverts :
                decouverts.append(v)
                f.enfile(v)
        

```

```python
# Tests 
parcours_largeur(G1, "A")

```

```python
def parcours_profondeur(G, s):
    """ Paramètres
    ---------
    G : (dict)
    un dictionnaire représentant un graphe
    
    s : (string)
    sommet de départ 
    
    Résultat
    --------
    None
    Cette fonction ne renvoie rien mais affiche les sommets du graphe au fur et à mesure qu'ils sont atteints 
    par le parcours en profondeur. 
    """
    p = Pile()
    p.empile(s)
    visites = []
    while p.est_vide() != True :
        s = p.depile()
        if s not in visites :
            print(s)
            visites.append(s)
            for v in voisins(G, s):
                if v not in visites :
                    p.empile(v)
    
```

```python
# Tests 
parcours_profondeur(G1, "A")
```

```python
def parcours_profondeur_recursif(G, s, visites):
    """ Paramètres
    ---------
    G : (dict)
    un dictionnaire représentant un graphe
    
    s : (string)
    sommet de départ 
    
    Résultat
    --------
    None
    Fonction récursive qui ne renvoie rien mais affiche les sommets du graphe au fur et à mesure qu'ils sont atteints 
    par le parcours en profondeur. 
    """
    visites.append(s)
    print(s)
    for v in voisins(G, s):
        if v not in visites :
            parcours_profondeur_recursif(G, v, visites)
```

### III - Détection d'un cycle dans un graphe

```python
def g_couleurs(G) :
    dico = {}
    for lettre in G :
        dico[lettre] = "BLANC"
    return dico
```

```python
def detection_cycle(G, s):
    """ Paramètre
    ---------
    G : (dict)
    un dictionnaire représentant un graphe
    
    s : (string)
    sommet quelconque du graphe G
        
    Résultat
    --------
    (bool)
    Cette fonction renvoie True si le graphe G contient un cycle, False sinon."""

    couleur = g_couleurs(G)

    p = Pile()
    p.empile(s)

    while p.est_vide() != True :
        u = p.depile()
        for v in G[u] :
            if couleur[v] != "NOIR" :
                p.empile(v)
        if couleur[u] == "NOIR" :
            return True
        else :
            couleur[u] = "NOIR"
    return False
```

### IV - Recherche d'un chemin dans un graphe

```python
def trouver_chemin(G, debut, fin, chaine):
    """ Paramètres
    ---------
    G : (dict)
    un dictionnaire représentant un graphe
    
    debut : (string)
    sommet de départ
    
    fin : (string)
    sommet d'arrivée
    
    Résultat
    --------
    (list)
    Cette fonction renvoie une liste de sommets correspondant à un chemin possible allant de debut à fin. 
    """
    
    chaine.append(debut)
    
    if debut == fin :
        return chaine
    for u in G[debut] :
        if u not in chaine :
            nchemin = trouver_chemin(G, u, fin, chaine)
            if len(nchemin) != 0 :
                return nchemin
    return None

```
