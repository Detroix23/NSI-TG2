### I - Définition des classes `Cellule`, `Pile` et `File`

Il n'y a rien d'autre à faire dans cette partie que d'exécuter les cellules de code : ces classes seront utiles dans la partie suivante pour implémenter les parcours en largeur et en profondeur d'un graphe. 


```python
class Cellule :
    def __init__(self, contenu, suivante):
        self.contenu = contenu
        self.suivante = suivante
```

```python
class Pile:
    def __init__(self):
        self.data = None
    
    def est_vide(self):
        return self.data == None
    
    def empile(self, x):
        self.data = Cellule(x,self.data)
    
    def depile(self):
        v = self.data.contenu #on récupère la valeur à renvoyer
        self.data = self.data.suivante  # on supprime la 1ère cellule  
        return v
    
    def __str__(self):
        s = ""
        c = self.data
        while c != None :
            s += "|\t" + str(c.contenu)+ "\t|" + "\n"
            c = c.suivante
        s = "\nEtat de la pile:\n" + s
        return s
   
```
