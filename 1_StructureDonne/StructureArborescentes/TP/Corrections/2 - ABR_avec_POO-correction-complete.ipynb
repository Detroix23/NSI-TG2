## 1. Implémentation d'un Arbre Binaire en utilisant la POO

### Ce qu'on veut faire

On veut créer une structure de données pour les arbres binaires d'entiers.  

Un arbre binaire d'entiers est ici :
- ou bien l'arbre vide ;
- ou bien un nœud contenant un entier et ses deux sous-arbres, gauche et droit, qui sont des arbres binaires d'entiers.

La définition est donc récursive.

L'idée est de créer une classe (qu'on pourra ranger dans un module), permettant les opérations suivantes :

- création d'un arbre vide ;
- création d'un arbre à partir d'un noeud et de deux sous-arbres ;
- test si un arbre est vide ou non ;
- racine d'un arbre ;
- sous-arbre gauche et sous-arbre droit d'un arbre. 



**Une classe pour les arbres binaires**  

On programme une classe `Noeud` (similaire à la classe `Cellule` de la partie sur les listes) et une classe `ArbreBinaire`.

````python
class Noeud:
    def __init__(self, valeur, gauche, droit):
        self.v = valeur
        self.g = gauche
        self.d = droit

class ArbreBinaire:
    def __init__(self, racine = None):
        self.racine = racine
           
    def creeNGD(valeur, gauche = None, droit = None):
        return ArbreBinaire(Noeud(valeur, gauche, droit))
    
    def estVide(self):
        return self.racine is None
    
    def get_racine(self):
        if self.estVide():
            raise ValueError("ABR vide")
        return self.racine.v
    
    def ssa_Gauche(self):
        if self.estVide():
            raise ValueError("ABR vide")
        return self.racine.g
    
    def ssa_Droit(self):
        if self.estVide():
            raise ValueError("ABR vide")
        return self.racine.d

vide = ArbreBinaire()

def creeFeuille(x):
    return ArbreBinaire.creeNGD(x, vide, vide)
````

On a défini une fonction **creeFeuille** pour créer un arbre réduit à un noeud-racine sans fils, pour simplifier l’utilisation de ces classes.
Il est alors facile de définir des fonctions **récursives** calculant la taille ou la hauteur d’un arbre binaire.

**On prendra comme convention que l'arbre vide a pour hauteur 0 (et donc un arbre réduit à une racine a une hauteur 1).**

````python
def taille(a):
    if a.estVide():
        return 0
    else:
        return 1 + taille(a.ssa_Gauche()) + taille(a.ssa_Droit())

def hauteur(a):
    if a.estVide():
        return 0
    else :
        return 1 + max(hauteur(a.ssa_Gauche()), hauteur(a.ssa_Droit()))
````

Voici un exemple d’utilisation et construisons l'arbre suivant :

![](data/Arbre_binaire.jpg)

````python
a1 = ArbreBinaire.creeNGD(4, ArbreBinaire.creeNGD(3, creeFeuille(1), vide), creeFeuille(6))
````

````python
a2 = ArbreBinaire.creeNGD(12, ArbreBinaire.creeNGD(9, vide, creeFeuille(11)), creeFeuille(14))
````

````python
a = ArbreBinaire.creeNGD(8, a1, a2)
````

On a ici construit l’arbre de droite de la figure qu’on a nommé **a**, en commençant par construire son sous-arbre gauche **a1** et son sous-arbre droit **a2**.

````python
assert taille(a) == 9
assert hauteur(a) == 4
assert a.ssa_Droit().ssa_Gauche().get_racine() == 9
assert a.ssa_Gauche().ssa_Gauche().ssa_Gauche().get_racine() == 1
````

## 2. Les Arbres Binaires de Recherche

Un **A**rbre **B**inaire de **R**echerche (**ABR**) est un arbre binaire qui vérifie de plus la propriété suivante : 

**les éléments qui figurent dans le sous-arbre gauche sont tous inférieurs ou égaux à la racine, ceux qui figurent dans le sous-arbre droit sont tous supérieurs ou égaux à la racine. En outre, chaque sous-arbre vérifie également cette propriété.**

Sur la base du paragraphe précédent, on peut définir une structure de données appelée Arbre Binaire de Recherche (ABR). 

Cette structure permet :

* la création d’une structure vide ;
* l’ajout d’un élément à la structure de données ;
* la recherche d’un élément, pour savoir s’il est présent ou non dans la structure.

On pourrait ajouter d’autres méthodes, l’une listant les éléments de la structure, l’autre supprimant un élément.  

La sémantique demande une petite précision : on peut ajouter un élément qui existe déjà, il sera alors présent deux fois dans la structure.

#### Voici un exemple d'ABR :

![](data/ABR.jpg)

L'arbre binaire de la partie 1 est également un Arbre Binaire de Recherche. 


**Ajout d’un élément dans un ABR**

Pour ajouter un élément dans un ABR, on procède comme suit :  
* si l'arbre est vide, on crée le noeud ayant un sous-arbre gauche et un sous-arbre droit vide ;  
* sinon :
  - si l'élément est inférieur à la racine, on cherche à l'ajouter récursivement dans le sous-arbre gauche ;
  - si l'élément est supérieur à la racine, on cherche à l'ajouter récursivement dans le sous-arbre droit ;
 
L'ajout se fait au premier emplacement vide trouvé qui respecte la propriété de l'arbre binaire de recherche. 

Voici la déclaration de la classe **ABR** qu’on obtient ainsi.

````python
class Noeud:
    def __init__(self, valeur, gauche = None, droit = None):
        self.v = valeur
        self.g = gauche
        self.d = droit

class ABR:
    def __init__(self, racine = None):
        self.racine = racine
    
    def estVide(self):
        return self.racine is None
    
    def get_racine(self):
        assert not self.estVide(), 'ABR vide'
        return self.racine.v
    
    def ssa_Gauche(self):
        assert not self.estVide(), 'ABR vide'
        return self.racine.g
    
    def ssa_Droit(self):
        assert not self.estVide(), 'ABR vide'
        return self.racine.d
        
    def ajoute(self, x):
        if self.estVide():
            self.racine = Noeud(x, ABR(), ABR())    # Création d'un ABR vide pour les sous arbres gauche et droit
        elif x <= self.get_racine():
            if self.ssa_Gauche() is None:
                self.racine.g = ABR(Noeud(x, ABR(), ABR()))
            else:
                self.ssa_Gauche().ajoute(x)
        else:
            if self.ssa_Droit() is None:
                self.racine.d = ABR(Noeud(x, ABR(), ABR()))
            else:
                self.ssa_Droit().ajoute(x)
````

On peut tester, en construisant l'arbre de la partie 1. 

````python
a = ABR()      # Création d'un Arbre Binaire de Recherche vide
a.ajoute(8)
a.ajoute(4)
a.ajoute(12)
a.ajoute(3)
a.ajoute(6)
a.ajoute(9)
a.ajoute(14)
a.ajoute(1)
a.ajoute(11)

a.ssa_Droit().ssa_Droit().get_racine() == 14
````

**La recherche d’un élément**

Cette propriété facilite la recherche d’un élément :  
* s’il est égal à la racine de l’arbre, on l’a trouvé ;  
* s’il est inférieur, on le recherche récursivement dans le fils gauche ;  
* s’il est supérieur, on le recherche récursivement dans le fils droit.  

Le nombre de comparaisons effectuées dans la recherche est donc borné par $h$ où $h$ est la hauteur de l’arbre. Il est donc de l’ordre du logarithme binaire de sa taille $n$, pour peu que l’arbre soit équilibré.  

On crée maintenant la classe **ABR** : 

````python
class Noeud:
    def __init__(self, valeur, gauche = None, droit = None):
        self.v = valeur
        self.g = gauche
        self.d = droit

class ABR:
    def __init__(self, racine = None):
        self.racine = racine
    
    def estVide(self):
        return self.racine is None
    
    def get_racine(self):
        if self.estVide():
            raise ValueError("ABR vide")
        return self.racine.v
    
    def ssa_Gauche(self):
        if self.estVide():
            raise ValueError("ABR vide")
        return self.racine.g
    
    def ssa_Droit(self):
        if self.estVide():
            raise ValueError("ABR vide")
        return self.racine.d
        
    def cherche(self, x):
        if self.estVide():
            return False
        elif x == self.get_racine():
            return True
        elif x < self.get_racine():
            return self.ssa_Gauche().cherche(x)
        else:
            return self.ssa_Droit().cherche(x)
            
    def ajoute(self, x):
        if self.estVide():
            self.racine = Noeud(x, ABR(), ABR())    # Création d'un ABR vide pour les sous arbres gauche et droit
        elif x <= self.get_racine():
            if self.ssa_Gauche() is None:
                self.racine.g = ABR(Noeud(x, ABR(), ABR()))
            else:
                self.ssa_Gauche().ajoute(x)
        else:
            if self.ssa_Droit() is None:
                self.racine.d = ABR(Noeud(x, ABR(), ABR()))
            else:
                self.ssa_Droit().ajoute(x)
````

````python
a = ABR()      # Création d'un Arbre Binaire de Recherche vide
a.ajoute(8)
a.ajoute(4)
a.ajoute(12)
a.ajoute(3)
a.ajoute(6)
a.ajoute(9)
a.ajoute(14)
a.ajoute(1)
a.ajoute(11)

a.ssa_Droit().ssa_Droit().get_racine() == 14
````

**Coût de la recherche dans un ABR équilibré**

On dit qu'un arbre binaire est équilibré si tous les chemins menant de la racine à une feuille ont pour longueur soit $h$, soit $h-1$, $h$ étant la hauteur de l'arbre. 

![](data/rechercheABR.png)

Imaginons un arbre équilibré de taille $n$. Combien d'étapes faudra-t-il, dans le pire des cas, pour trouver (ou pas) une clé particulière dans cet arbre ?

Après chaque nœud, le nombre de nœuds restant à explorer est divisé par 2. 


S'il faut parcourir tous les étages de l'arbre avant de trouver (ou pas) la clé recherchée, le nombre de nœuds parcourus est donc égal à la hauteur $h$ de l'arbre.

Pour un arbre complet, cette hauteur vérifie la relation $2^h -1= n$. et donc $2^h = n+1$.

$h$ est donc le «nombre de puissances de 2» que l'on peut mettre dans $n+1$. Cette notion s'appelle le logarithme de base 2 et se note $\log_2$.

Par exemple, $\log_2(64)=6$ car $2^6=64$.

Le nombre maximal de nœuds à parcourir pour rechercher une clé dans un ABR équilibré de taille $n$ est donc de l'ordre de $\log_2(n)$, ce qui est très performant !

Pour un arbre contenant 1000 valeurs, 10 étapes suffisent.

Cette **complexité logarithmique** est un atout essentiel de la structure d'arbre binaire de recherche.

## 3. Autres algorithmes sur les arbres

### 3.1 Déterminer si un arbre est un ABR

Dans la classe `ArbreBinaire` définie au début de ce notebook, on a ajouté la méthode `estABR` qui vaut `True` si l'arbre binaire est un arbre binaire de recherche, et `False` sinon. 

**Question :**  
Analyser et expliquer le script de cette méthode. 

````python
class Noeud:
    def __init__(self, valeur, gauche, droit):
        self.v = valeur
        self.g = gauche
        self.d = droit

class ArbreBinaire:
    def __init__(self, racine = None):
        self.racine = racine
           
    def creeNGD(valeur, gauche = None, droit = None):
        return ArbreBinaire(Noeud(valeur, gauche, droit))
    
    def estVide(self):
        return self.racine is None
    
    def get_racine(self):
        assert not(self.racine is None), 'Arbre vide'
        return self.racine.v
    
    def ssa_Gauche(self):
        assert not(self.racine is None), 'Arbre vide'
        return self.racine.g
    
    def ssa_Droit(self):
        assert not(self.racine is None), 'Arbre vide'
        return self.racine.d

    def estABR(self, min_val = float('-inf'), max_val = float('inf')):
        """Vérifie si l'arbre est un ABR."""
        if self.estVide():
            return True
        if self.get_racine() <= min_val or self.get_racine() >= max_val:
            return False
        return (self.ssa_Gauche().estABR(min_val, self.get_racine())) and (self.ssa_Droit().estABR(self.get_racine(), max_val))

vide = ArbreBinaire()

def creeFeuille(x):
    return ArbreBinaire.creeNGD(x, vide, vide)
````

````python
# Tests
a1 = ArbreBinaire.creeNGD(4, creeFeuille(7), creeFeuille(1))
a2 = ArbreBinaire.creeNGD(12, ArbreBinaire.creeNGD(9, vide, creeFeuille(11)), creeFeuille(14))
a = ArbreBinaire.creeNGD(8, a1, a2)

assert a1.estABR() == False
assert a2.estABR() == True
assert a.estABR() == False
````

### 3.2 Calcul du nombre de feuilles d'un arbre

**Rappel :**  
Une feuille est un nœud d'arité 0, autrement dit sans fils gauche, ni fils droit.

````python
def nbfeuilles(arbre):
    ag = arbre.ssa_Gauche()
    ad = arbre.ssa_Droit()
    if ag.estVide() and ad.estVide():
        return 1
    elif ag.estVide():
        return nbfeuilles(ad)
    elif ad.estVide():
        return nbfeuilles(ag)
    else:
        return nbfeuilles(ag) + nbfeuilles(ad)
    
````
