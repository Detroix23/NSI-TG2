## 1. Rappels de vocabulaire  
Un arbre est une structure hiérarchique de données, composée de nœuds.  

Si on adopte le vocabulaire des graphes (qui seront vus plus tard dans l'année), un arbre est un graphe non orienté, connexe, sans cycle, et dans lequel un nœud joue le rôle de racine.

![](data/term.png) 


- Chaque **nœud** a exactement un seul **nœud père**, à l'exception du nœud **racine** qui est le seul nœud à ne pas avoir de père. (oui, **la** racine d'une arbre est **en haut**)
<img src="data/real_tree.png" width='30%' />
- Chaque nœud peut avoir un nombre quelconque de **fils**, dont il est le père.
- Les nœuds qui n'ont pas de fils sont appelés les **feuilles** (ou nœuds externes).
- Les nœuds qui ne sont pas des feuilles sont des **nœuds internes**.
- Le nom de chaque nœud est appelé son **étiquette**.

**Exemples :**
dans l'arbre ci-dessus,
- C est la racine, E, Z A et G sont les feuilles.
- K est le père de A et G.
- F est le père de Z.
- C est le père de B et K
- B est le père de E et F.

## 2. Parcours d'arbres
Les arbres étant une structure hiérarchique, leur utilisation implique la nécessité d'un **parcours** des valeurs stockées. Par exemple pour toutes les récupérer dans un certain ordre, ou bien pour en chercher une en particulier.  

Il existe plusieurs manières de parcourir un arbre.


### 2.1 Parcours en largeur d'abord (BFS)
*BFS : Breadth First Search*

Le parcours en largeur d'abord est un parcours étage par étage (de haut en bas) et de gauche à droite.

![](data/BFS.png)

L'ordre des lettres parcourues est donc T-Y-O-P-H-N.

Les trois parcours que nous allons voir maintenant sont des parcours en **profondeur d'abord**, ou **DPS** (*Depth First Search*). Ce qui signifie qu'un des deux sous-arbres sera totalement parcouru avant que l'exploration du deuxième ne commence. 

### 2.2 Parcours préfixe
Le parcours **préfixe** est un parcours **en profondeur d'abord**. 

**Méthode du parcours préfixe :** (parfois aussi appelé *préordre*)
- Chaque nœud est visité avant que ses fils le soient.
- On part de la racine, puis on visite son fils gauche (et éventuellement le fils gauche de celui-ci, etc.) avant de remonter et de redescendre vers le fils droit.

![](data/prefixe.png)

L'ordre des lettres parcourues est donc T-Y-P-O-H-N.

### 2.3 Parcours infixe
Le parcours **infixe** est aussi un parcours en profondeur d'abord.

**Méthode du parcours infixe :** (parfois aussi appelé *en ordre*)
- Chaque nœud est visité **après son fils gauche mais avant son fils droit**.
- On part donc de la feuille la plus à gauche et on remonte par vagues sucessives. Un nœud ne peut pas être visité si son fils gauche ne l'a pas été.

![](data/infixe.png)

L'ordre des lettres parcourues est donc P-Y-T-H-O-N.

### 2.4 Parcours postfixe
Le parcours **postfixe** est aussi un parcours en profondeur d'abord.

**Méthode du parcours postfixe :** (parfois aussi appelé *post ordre*)
- Chaque nœud est visité **après ses fils le soient**.
- On part donc de la feuille la plus à gauche, et on ne remonte à un nœud père que si ses fils ont tous été visités. 

![](data/postfixe.png)

L'ordre des lettres parcourues est donc P-Y-H-N-O-T.

### 2.5 Comment ne pas se mélanger entre le pré / in / post fixe ?
- *pré* veut dire *avant*
- *in* veut dire *au milieu*
- *post* veut dire *après*

Ces trois mots-clés parlent de la place du **père** par rapport à ses fils. 
Ensuite, il faut toujours se souvenir qu'on traite le fils gauche avant le fils droit.

- préfixe : le père doit être le premier par rapport à ses fils.
- infixe : le père doit être entre son fils gauche (traité en premier) et son fils droit.
- postfixe : le père ne doit être traité que quand ses deux fils (gauche d'abord, droite ensuite) l'ont été.

Un parcours préfixe commencera toujours par la racine, alors qu'un parcours postfixe finira toujours par la racine. Dans un parcours infixe, la racine sera «au milieu» (pas nécessairement parfaitement).

### 3 Parcours en profondeur : préfixe, infixe, postfixe - implémentation

On définit à nouveau la classe la classe `ABR` déjà rencontrées dans le Notebook *2- ABR_avec_POO.ipynb*.  


````python
class Noeud:
    def __init__(self, valeur, gauche = None, droit = None):
        self.v = valeur
        self.g = gauche
        self.d = droit

class ABR:
    def __init__(self, racine = None):
        self.racine = racine
    
    def estVide(self):
        return self.racine is None
    
    def get_racine(self):
        assert not self.estVide(), 'ABR vide'
        return self.racine.v
    
    def ssa_Gauche(self):
        assert not self.estVide(), 'ABR vide'
        return self.racine.g
    
    def ssa_Droit(self):
        assert not self.estVide(), 'ABR vide'
        return self.racine.d
        
    def cherche(self, x):
        if self.estVide():
            return False
        elif x == self.get_racine():
            return True
        elif x < self.get_racine():
            if self.ssa_Gauche() is None:
                return False
            else:
                return self.ssa_Gauche().cherche(x)
        else:
            if self.ssa_Droit() is None:
                return False
            else:
                return self.ssa_Droit().cherche(x)
        
    def ajoute(self, x):
        if self.estVide():
            self.racine = Noeud(x, ABR(), ABR()) #Création d'un ABR vide pour les sous arbres gauche et droit
        elif x <= self.get_racine():
            if self.ssa_Gauche() is None:
                self.racine.g = ABR(Noeud(x, ABR(), ABR()))
            else:
                self.ssa_Gauche().ajoute(x)
        else:
            if self.ssa_Droit() is None:
                self.racine.d = ABR(Noeud(x, ABR(), ABR()))
            else:
                self.ssa_Droit().ajoute(x)
````

Pour tester les fonctions `prefixe`, `infixe` et `postfixe` qui seront définies ci-après, on crée l'arbre binaire de recherche `a` suivant :

````python
a = ABR()      # Création d'un Arbre Binaire de Recherche vide
a.ajoute(8)
a.ajoute(4)
a.ajoute(12)
a.ajoute(3)
a.ajoute(6)
a.ajoute(9)
a.ajoute(14)
a.ajoute(1)
a.ajoute(11)
````

### 4 Parcours en largeur
Le parcours en largeur (BFS) est le plus simple à faire visuellement mais il est plus difficile à coder que les parcours préfixe, infixe, postfixe.  
Il est nécessaire d'utiliser une **file**  :
- On place l'arbre dans la file.
- Tant que la file n'est pas vide, on procède comme suit :
    - On défile, donc on récupère l'arbre situé en haut de la file.
    - On garde son étiquette.
    - On enfile, s'il n'est pas vide, son sous-arbre gauche
    - On enfile, s'il n'est pas vide, son sous-arbre droit.
    
<center> <img src='data/ABRv3.jpg' width='300'> </center>

![](data/parcoursBFS.png)

On définit ci-dessous la classe `File`, déjà rencontrée dans le cours sur les structures linéaires. Pour rappel, elle permet de : 
- créer une file vide avec ```file = File()```
- défiler un élément par ```file.defile()```
- enfiler l'élément ```a``` par ```file.enfile(a)```
- savoir si la file est vide par le booléen ```file.est_vide()```

````python
class Cellule :
    def __init__(self, contenu, suivante):
        self.contenu = contenu
        self.suivante = suivante
        
class File:
    def __init__(self):
        self.data = None
    
    def est_vide(self):
        return self.data == None
    
    def enfile(self, x):
        self.data = Cellule(x, self.data)
    
    def defile(self):
        c = self.data
        if c == None:
            raise IndexError("Vous avez essayé de défiler une file vide !")
        elif c.suivante != None:    #si la liste chaînée ne contient pas qu'un seul élément
            while c.suivante != None:
                prec = c
                c = c.suivante
            prec.suivante = None
        else:                     #Cas où la liste chaînée contient un seul élément
            self.data = None
        return c.contenu
````

Ecrire, dans la cellule ci-dessous, le code de la fonction `BFS` qui prend en paramètre un arbre binaire `arbre` et qui affiche les étiquettes des noeuds lorsqu'on parcourt l'arbre en largeur. 

````python
def BFS(arbre):
    file = File()      # Création d'une file vide
    file.enfile(arbre)
    while not file.est_vide():
        arbre_defile = file.defile()
        print(arbre_defile.get_racine(), end = '-')
        if not arbre_defile.ssa_Gauche().estVide():
            file.enfile(arbre_defile.ssa_Gauche())
        if not arbre_defile.ssa_Droit().estVide():
            file.enfile(arbre_defile.ssa_Droit())
````

#### 5.1 Déterminer si un arbre est un ABR

La fonction `est_ABR` suivante prend en paramètre un arbre binaire `arbre` et retourne un booléen qui vaut `True` si `arbre` est un arbre binaire de recherche (ou l'arbre vide), et `False` sinon. 

````python
def est_ABR(arbre):
    """
    Paramètre
    ---------
    arbre : (ABR)
    un Arbre Binaire de Recherche
    
    Résultat
    --------
    (booléen)
    Cette fonction renvoie True si arbre est un ABR, False sinon.
    """
    if arbre.estVide():
        # L'arbre est vide
        return True
    
    elif (arbre.ssa_Gauche().estVide()) and (arbre.ssa_Droit().estVide()):
        # L'arbre est réduit à une feuille
        return True
    
    elif arbre.ssa_Droit().estVide():
        # Arbre avec une racine et seulement un sous-arbre gauche
        if arbre.ssa_Gauche().get_racine() <= arbre.get_racine():
            return est_ABR(arbre.ssa_Gauche())
        else:
            return False
        
    elif arbre.ssa_Gauche().estVide():
        # Arbre avec une racine et seulement un sous-arbre droit
        if arbre.ssa_Droit().get_racine() >= arbre.get_racine():
            return est_ABR(arbre.ssa_Droit())
        else:
            return False
    else:
        # Arbre ayant une racine, un sous-arbre gauche et un sous-arbre droit
        if arbre.ssa_Gauche().get_racine() <= arbre.get_racine():
            return est_ABR(arbre.ssa_Gauche())
        elif arbre.ssa_Droit().get_racine() >= arbre.get_racine():
            return est_ABR(arbre.ssa_Droit())
        else:
            return False

est_ABR(a)
````
