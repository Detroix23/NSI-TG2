### I - Implémentation d'une classe pour les arbres binaires

Les méthodes, fonctions et variables définies dans cette partie pourront être utilisées sans aucune modification. 

On programme une classe `Noeud` (similaire à la classe `Cellule` de la partie sur les listes) et une classe `ArbreBinaire`.

```python
class Noeud:
    def __init__(self,valeur,gauche,droit):
        self.v = valeur
        self.g = gauche
        self.d = droit

class ArbreBinaire:
    def __init__(self,a):
        self.a = a
        
    def creeVide():
        return ArbreBinaire(None)
    
    def creeNGD(valeur,gauche=None,droit=None):
        return ArbreBinaire(Noeud(valeur,gauche,droit))
    
    def estVide(self):
        return self.a is None
    
    def racine(self):
        assert not(self.a is None),'Arbre vide'
        return self.a.v
    
    def filsGauche(self):
        assert not(self.a is None),'Arbre vide'
        return self.a.g
    
    def filsDroit(self):
        assert not(self.a is None),'Arbre vide'
        return self.a.d

vide = ArbreBinaire(None)  #Arbre vide

def creeFeuille(x):
    return ArbreBinaire.creeNGD(x, vide, vide)
```

**On prendra comme convention que l'arbre vide a pour hauteur 0 (et donc un arbre réduit à une racine a une hauteur 1).**

```python
def taille(a):
    if a.estVide():
        return 0
    else:
        return 1 + taille(a.filsGauche()) + taille(a.filsDroit())

def hauteur(a):
    if a.estVide():
        return 0
    else :
        return 1 + max(hauteur(a.filsGauche()), hauteur(a.filsDroit()))
```

```python
#Définition des fonctions racine, gauche, droit et est_vide mentionnés dans l'énoncé

def racine(arb):
    """Cette fonction prend en paramètre un arbre de compétition arb et renvoie la valeur de la racine"""
    return arb.racine()

def gauche(arb):
    """Cette fonction prend en paramètre un arbre de compétition arb et renvoie son sous-arbre gauche."""
    return arb.filsGauche()

def droit(arb):
    """Cette fonction prend en paramètre un arbre de compétition arb et renvoie son sous-arbre droit."""
    return arb.filsDroit()

def est_vide(arb):
    """Cette fonction prend en paramètre un arbre de compétition et renvoie True si l’arbre est vide et False sinon."""
    return taille(arb) == 0
```

Construction des arbres de l'énoncé, et d'un troisième arbre réduit à un seul joueur. 

```python
# Construction de l'arbre A
a = ArbreBinaire.creeNGD("Kamel", ArbreBinaire.creeNGD("Kamel", creeFeuille("Joris"), creeFeuille("Kamel")), ArbreBinaire.creeNGD("Carine", creeFeuille("Carine"), creeFeuille("Abdou")))

# Construction de l'arbre B
b1 = ArbreBinaire.creeNGD("Léa", ArbreBinaire.creeNGD("Léa", creeFeuille("Marc"), creeFeuille("Léa")), ArbreBinaire.creeNGD("Théo", creeFeuille("Claire"), creeFeuille("Théo"))) 
b2 = ArbreBinaire.creeNGD("Louis", ArbreBinaire.creeNGD("Louis", creeFeuille("Marie"), creeFeuille("Louis")), ArbreBinaire.creeNGD("Anne", creeFeuille("Anne"), creeFeuille("Kévin")))
b = ArbreBinaire.creeNGD("Léa",b1,b2)

# Construction de l'arbre C réduit à un seul joueur. 
c = creeFeuille("David")
```

```python
#Exemples d'utilisation 
print(racine(a))
print(racine(droit(a)))

print(est_vide(a))
```
