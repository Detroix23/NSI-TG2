### Exercice 1: parcours d'une liste, création d'un dictionnaire
    

Le gestionnaire d'un site de ventes en ligne stocke l’ensemble des ventes successives dans une liste qui contient des nombres entiers.  
Chaque nombre entier fait référence à un article vendu sur le site.  

Afin de gérer ce site, le gestionnaire crée une fonction `comptage` qui :  
* prend en paramètre une liste de produits vendus, identifiés par un numéro ;
* renvoie le dictionnaire où chaque nombre entier référençant un article vendu est une clé et où le nombre de fois où cet article a été vendu est la valeur associée à cette clé.

Par exemple :

```python
>>> ventes = [15, 102, 64, 15, 15, 15, 3, 64, 7, 32, 97, 64, 15]
>>> comptage(ventes) 
{15: 5, 102: 1, 64: 3, 3: 1, 7: 1, 32: 1, 97: 1}
>>> comptage([])
{}
```

Écrire le script de la fonction `comptage`.  


````python
def comptage(liste):
    dico = {}
    for nb in liste:
        if nb not in dico:
            dico[nb] = 1
        else:
            dico[nb] += 1
            
    return dico


comptage([15, 102, 64, 15, 15, 15, 3, 64, 7, 32, 97, 64, 15])
````

### Exercice 2 : parcours d'un dictionnaire, création d'une liste
Vous travaillez comme développeur dans une entreprise publicitaire et vous avez à déterminer les jours de la semaine où le temps d’attente moyen au téléphone des “clients importants” dépasse 12 secondes.  

Le dictionnaire `dis_allo` suivant associe, à chaque jour ouvré de la semaine, la liste des temps d’attente au téléphone connus par les “clients importants” ce jour-là.  

```python
dis_allo = {'lundi': [5, 3],
            'mardi': [18, 6, 8],
            'mercredi': [9, 11, 4, 6],
            'jeudi': [7],
            'vendredi': [2, 17, 19, 15],
            'samedi': [8, 13, 9, 6, 9]}
```
Écrire une fonction `liste_depassements` qui prend en paramètre un dictionnaire `dico` d’associations jour / liste de temps d’attente et qui renvoie la liste des jours où le temps d'attente moyen des “clients importants” dépasse 12 secondes. 

Par exemple :
```python
>>> liste_depassements(dis_allo)
['mardi', 'vendredi']
>>> liste_depassements({'lundi': [6], 'mardi': [5, 9, 1], 'mercredi': [10], 'jeudi': [], 'vendredi': [4, 4],'samedi':[6, 11, 11]})
[]
```

````python
def moyenne(liste):
    s = 0
    for elem in liste:
        s += elem
    return s / len(liste)

def liste_depassements(dico):
    depass = []
    for jour, liste in dico.items():
        if (liste != []) and (moyenne(liste) > 12) and (jour not in depass):
             depass.append(jour)
                
    return depass
````

````python
dis_allo = {'lundi': [5, 3],
            'mardi': [18, 16, 8],
            'mercredi': [9, 11, 4, 6],
            'jeudi': [7],
            'vendredi': [2, 17, 19, 15],
            'samedi': [8, 13, 9, 6, 9]}
assert liste_depassements(dis_allo) == ['mardi', 'vendredi']
assert liste_depassements({'lundi': [6], 'mardi': [5, 9, 1], 'mercredi': [10], 'jeudi': [], 'vendredi': [4, 4],'samedi':[6, 11, 11]}) == []
````
