Programmer la fonction `est_en_ordre_decroissant` qui prend en paramètre un tableau de valeurs numériques non vide et qui renvoie `True` si ce tableau est trié dans l’ordre décroissant, `False` sinon. Cette fonction devra réussir les tests suivants : 

```python
assert est_en_ordre_decroissant([9, 8, 8, 8, 0]) == True
```

```python
assert est_en_ordre_decroissant([4, 12, 8]) == False 
```

```python
assert est_en_ordre_decroissant([4, -1]) == True
```

```python
assert est_en_ordre_decroissant([7]) == True
```

```python
def est_en_ordre_decroissant(L):
    for i in range (len(L)-1):
        if L[i]<L[i+1]:
            return False
    return True



```
