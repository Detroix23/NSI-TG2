```python
# Les instructions suivantes permettent de générer aléatoirement une pyramide de 25 niveaux
from random import randint

ny = 25
P = []
for y in range(ny):
    A=[]
    for x in range(y+1):
        A.append(randint(0,100))
    P.append(A)
print(P)
```

```python
def raisonnable(P):
    n = len(P)
    x = 0
    y = 0
    C = [P[0][0]]
    
    while y < n - 1:
        y = y + 1     # Passage à l'étage suivant
        if P[y][x] < P[y][x + 1]:
            x = x + 1
        C.append(P[y][x])
               
    return C
```

```python
# force brute récursif
def somme_max(P, x, y):
    if y == len(P)-1:
        return P[y][x]
    else:
        return P[y][x] + max(somme_max(P, x, y + 1), somme_max(P, x + 1, y + 1))
```

## Qu'est ce que la programmation dynamique ?

1. Le problème peut être résolu à partir de sous-problèmes similaires mais plus petits ;

2. L’ensemble de ces sous-problèmes est discret, c’est-à-dire qu’on peut les indexer et ranger les résultats dans un tableau ;

3. La solution optimale au problème posé s’obtient à partir des solutions optimales des sous-problèmes ;

4. Les sous-problèmes ne sont pas indépendants et un traitement récursif fait apparaître les mêmes sous-problèmes un grand nombre de fois.

```python
# algorithme programmation dynamique
def somme_max_dyn(P):
    n=len(P)
    memo=[[0]*(i+1) for i in range(n)]
    for x in range(n):
        memo[n-1][x]=P[n-1][x]
    y=n-1
    while y>0:
        for x in range(y):
            memo[y-1][x]=P[y-1][x]+max(memo[y][x],memo[y][x+1])
        y=y-1
        
    return memo
    # return memo[0][0]
```

```python
somme_max_dyn(P)
```

```python
# reconstitution de la solution
def solution(P):
    memo = somme_max_dyn(P)
     
    n = len(P)
    sol = []
    x = 0
    for y in range(n):
        sol.append(P[y][x])
        if y < n-1:
            if memo[y + 1][x] < memo[ y + 1][x + 1]:
                x = x + 1
    return sol            
```
